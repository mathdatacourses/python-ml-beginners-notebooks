In this lecture, we will use several Pandas operations together to answer our guiding question:

**Which students studied at least 5 hours and scored at least 80?**

## Setup

Import Pandas and load the final student dataset into a DataFrame called `df`.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/final_student_dataset.csv"

df = pd.read_csv(url)

df
```

## Select the Columns We Need

Our question uses:

- `Name`
- `StudyHours`
- `Score`

The `Attendance` column is not needed for this question.

```python
df_selected = df[["Name", "StudyHours", "Score"]]

df_selected
```

## Missing Values and Our Question

Two rows contain missing values:

- Noah is missing `Score`.
- Emma is missing `StudyHours`.

However, we can still determine that neither student satisfies both conditions.

Noah studied only 3.5 hours, so he does not satisfy `StudyHours >= 5`.

Emma scored 79, so she does not satisfy `Score >= 80`.

Therefore, neither student can satisfy both conditions.

## Create a Complete Working DataFrame

For the rest of this analysis, we will create a DataFrame that contains complete rows for the selected columns.

We use `dropna()` to remove rows that contain a missing value.

The original DataFrame `df` is unchanged.

```python
df_clean = df_selected.dropna()

df_clean
```

`df_clean` contains 6 rows.

For this particular question, removing Noah and Emma was not necessary to determine that they fail the conditions.

However, `dropna()` gives us a complete working DataFrame and reviews an important data-cleaning operation.

## Filter Using Two Conditions

We want students who satisfy both:

- `StudyHours >= 5`
- `Score >= 80`

Because both conditions must be true, we combine them using `&`.

Each condition must be inside parentheses.

```python
result = df_clean[
    (df_clean["StudyHours"] >= 5) &
    (df_clean["Score"] >= 80)
]

result
```

The students who satisfy both conditions are:

- Ava
- Maya
- Zoe
- Lucas

## Sort the Result

Sort the matching students by `Score` from highest to lowest.

```python
result = result.sort_values("Score", ascending=False)

result
```

## Putting the Workflow Together

The complete workflow is:

```python
df_selected = df[["Name", "StudyHours", "Score"]]

df_clean = df_selected.dropna()

result = df_clean[
    (df_clean["StudyHours"] >= 5) &
    (df_clean["Score"] >= 80)
]

result = result.sort_values("Score", ascending=False)

result
```

A useful Pandas workflow is:
