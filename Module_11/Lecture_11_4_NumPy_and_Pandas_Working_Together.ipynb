## Setup

Import Pandas and NumPy, load the final student dataset, and create a DataFrame containing only complete rows.

```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/final_student_dataset.csv"

df = pd.read_csv(url)
df_clean = df.dropna()

df_clean
```

## Select the Score Column

Select the `Score` column from `df_clean`.

The result is a Pandas Series.

```python
df_clean["Score"]
```

## Convert the Pandas Series to a NumPy Array

Use `to_numpy()` to convert the Score values to a NumPy array.

```python
scores = df_clean["Score"].to_numpy()

scores
```

## Perform a NumPy Operation

Suppose 5 points are added to every Score.

NumPy performs the addition on every element of the array.

```python
adjusted_scores = scores + 5

adjusted_scores
```

The original `scores` array has not changed.

```python
scores
```

## Create a Boolean Mask

Create a Boolean array that identifies Scores greater than or equal to 85.

```python
mask = scores >= 85

mask
```

Each Boolean value corresponds to one row of `df_clean`.

## Use the Mask with the DataFrame

Use the Boolean mask to select rows from `df_clean`.

```python
df_clean[mask]
```

The result is a DataFrame whose rows are the students whose Scores are at least 85.

These students are Ava, Zoe, and Lucas.

## Find the Minimum and Maximum

Use NumPy to find the smallest and largest Scores.

```python
np.min(scores)
```

```python
np.max(scores)
```

For the complete rows:

- the minimum Score is 63,
- the maximum Score is 91.

## Putting the Workflow Together

The complete workflow is:

```python
df_clean = df.dropna()

scores = df_clean["Score"].to_numpy()

adjusted_scores = scores + 5

mask = scores >= 85

df_clean[mask]

np.min(scores)
np.max(scores)
```

## What You Should Remember

Pandas and NumPy can work together in the same data workflow.

- Pandas is useful for organizing and selecting labeled tabular data.
- NumPy is useful for numerical arrays and numerical operations.
- `to_numpy()` converts a Pandas Series to a NumPy array.
- NumPy can perform element-by-element numerical operations.
- A NumPy Boolean array can be used to select rows from a Pandas DataFrame.
