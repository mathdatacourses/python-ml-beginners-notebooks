## Setup

Import Pandas and Matplotlib, load the final student dataset, and create a DataFrame containing only complete rows.

```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/final_student_dataset.csv"

df = pd.read_csv(url)
df_clean = df.dropna()

df_clean
```

## Which Plot Should We Use?

We want to examine the relationship between:

- `StudyHours`
- `Score`

Both variables are numerical.

A scatter plot is an appropriate choice.

## Create the Scatter Plot

Use `StudyHours` for the horizontal coordinates and `Score` for the vertical coordinates.

```python
plt.scatter(
    df_clean["StudyHours"],
    df_clean["Score"],
    s=80
)

plt.show()
```

## Put the Complete Plot Together

Add the figure size, axis labels, and title.

```python
plt.figure(figsize=(4, 3))

plt.scatter(
    df_clean["StudyHours"],
    df_clean["Score"],
    s=80
)

plt.xlabel("Study Hours", fontsize=12)
plt.ylabel("Score", fontsize=12)

plt.title("Score versus Study Hours", fontsize=14)

plt.show()
```

## Interpreting the Scatter Plot

Each point represents one student.

- The horizontal position represents the student's weekly StudyHours.
- The vertical position represents the student's Score.

Because this dataset is very small, we should use the plot for visual exploration and practice rather than drawing broad conclusions from the pattern.
