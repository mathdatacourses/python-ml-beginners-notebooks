# Lecture 11.2 — Loading and Inspecting the Data

In this lecture, we will load the final student dataset and inspect it before making any changes.


## Load the Dataset

Import Pandas and load the CSV file into a DataFrame called `df`.


````python
import pandas as pd

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/final_student_dataset.csv"

df = pd.read_csv(url)
````

## Look at the First Rows

Use `head()` to display the first five rows.


````python
df.head()
````

## Check the Shape

The `shape` attribute gives the number of rows and columns.


````python
df.shape
````

The result should be:

`(8, 4)`

This means the DataFrame has 8 rows and 4 columns.


## Check the Column Names


````python
df.columns
````

The four columns are:

- `Name`
- `StudyHours`
- `Attendance`
- `Score`


## Check the Data Types


````python
df.dtypes
````

Notice that `StudyHours` and `Score` are stored as `float64`.

Each of these columns contains a missing value.


## Get More Information

Use `info()` to see a compact summary of the DataFrame.


````python
df.info()
````

Look at the number of non-null values.

The DataFrame has 8 rows, but `StudyHours` and `Score` each have only 7 non-null values.


## Count Missing Values Directly

Use `isna().sum()` to count the missing values in each column.


````python
df.isna().sum()
````

The result confirms:

- `Name` has 0 missing values.
- `StudyHours` has 1 missing value.
- `Attendance` has 0 missing values.
- `Score` has 1 missing value.


## A Useful Inspection Routine

When you begin working with a new dataset, these commands provide a useful first inspection:

```python
df.head()
df.shape
df.columns
df.dtypes
df.info()
df.isna().sum()
```
