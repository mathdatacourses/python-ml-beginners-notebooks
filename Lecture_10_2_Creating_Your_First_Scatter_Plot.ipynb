## Import Matplotlib

We will use `matplotlib.pyplot` and give it the shorter name `plt`.


````python
import matplotlib.pyplot as plt
````

## The Data

- `StudyHours` provides the horizontal coordinates.
- `Score` provides the vertical coordinates.


````python
StudyHours = [6.5, 4.0, 2.5, 7.0]
Score = [88, 74, 63, 91]
````

## Create the Scatter Plot

The first argument gives the horizontal coordinates. The second gives the vertical coordinates.


````python
plt.scatter(StudyHours, Score)
````

## Display the Figure


````python
plt.show()
````

## Complete Example


````python
plt.scatter(StudyHours, Score)
plt.show()
````

## Axis Limits
