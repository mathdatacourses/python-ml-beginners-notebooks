## Import NumPy

````python
import numpy as np
````

## Selecting Values from an Array

Once we have created a NumPy array, we often want to select:

- one element,
- several consecutive elements,
- a row,
- a column,
- part of a two-dimensional array.

NumPy gives us tools for all of these tasks.

## Indexing a One-Dimensional Array

Suppose:

```python
scores = np.array([82, 91, 74, 88])
```

To select one element, we use its index.

````python
scores = np.array([82, 91, 74, 88])
````

````python
scores[0]
````

NumPy uses **zero-based indexing**, just like Python lists.

So:

```python
scores[0]
```

selects 82.

````python
print(scores[1])
````

````python
print(scores[3])
````

When Colab displays a NumPy scalar directly, it may sometimes also show its NumPy data type.

Using `print()` displays just the value.

## Negative Indices

Negative indices count backward from the end.

- `scores[-1]` is the last element.
- `scores[-2]` is the second-to-last element.

````python
print(scores[-1])
````

````python
print(scores[-2])
````

## Slicing a One-Dimensional Array

A slice selects several consecutive elements.

The general form is:

```python
array[start:stop]
```

The start index is included.

The stop index is not included.

````python
scores[1:3]
````

For this array, `scores[1:3]` returns:

```python
array([91, 74])
```

This is the same start-included, stop-not-included rule we saw with `range()` and `np.arange()`.

## Leaving Out Start or Stop

We can omit the starting index:

```python
scores[:2]
```

or omit the stopping index:

```python
scores[2:]
```

````python
scores[:2]
````

````python
scores[2:]
````

## A Two-Dimensional Array

Now consider:

```python
data = np.array([[10, 20, 30],
                 [40, 50, 60]])
```

This array has 2 rows and 3 columns.

````python
data = np.array([[10, 20, 30],
                 [40, 50, 60]])
````

````python
data
````

## Selecting One Element

For a two-dimensional array, we specify:

```python
array[row, column]
```

Remember: **row first, column second**.

````python
print(data[0, 1])
````

````python
print(data[1, 2])
````

Both row indices and column indices start at zero.

## Selecting an Entire Row

To select row 0:

```python
data[0, :]
```

The colon in the column position means **all columns**.

````python
data[0, :]
````

## Selecting an Entire Column

To select column 1:

```python
data[:, 1]
```

The colon in the row position means **all rows**.

````python
data[:, 1]
````

## Selecting Part of a Two-Dimensional Array

Consider a larger array:

````python
data = np.array([[10, 20, 30],
                 [40, 50, 60],
                 [70, 80, 90]])
````

We can select a rectangular section using:

```python
data[0:2, 1:3]
```

The first slice selects rows 0 and 1.

The second slice selects columns 1 and 2.

````python
data[0:2, 1:3]
````

The result is:

```python
array([[20, 30],
       [50, 60]])
```

## Row First, Column Second

For a two-dimensional array:

```python
array[row, column]
```

For slices:

```python
array[row_slice, column_slice]
```

A useful phrase to remember is:

**row first, column second**.

## What You Should Remember

For a one-dimensional array:

```python
array[index]
```

selects one element, and:

```python
array[start:stop]
```

selects a slice.

For a two-dimensional array:

```python
array[row, column]
```

selects one element, and:

```python
array[row_slice, column_slice]
```

selects part of the array.

Remember: **row first, column second**.
