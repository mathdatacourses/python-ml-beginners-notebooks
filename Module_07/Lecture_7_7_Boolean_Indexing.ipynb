## Selecting Values Using a Condition

So far, we have selected array elements using indices, slices, rows, and columns.

Sometimes we want to select values based on a condition.

For example, we may want to select all scores that are at least 80.

NumPy makes this easy using **Boolean indexing**.

## Plan

We will learn what a mask of a one-dimensional array is, how to use a mask to select elements, how a condition creates a mask, how to select the elements that satisfy a condition, and how to combine two conditions using element-by-element and and or.

## Masks of a One-Dimensional Array

A mask of a one-dimensional array is a one-dimensional Boolean array with the same length.

A Boolean array is an array whose elements are Boolean values.

````python
scores = np.array([82, 91, 74, 88, 65])

boolean_array_1 = np.array([True, True, False, True, False])
boolean_array_2 = np.array([True, True, False, False])
````

Both `boolean_array_1` and `boolean_array_2` are Boolean arrays.

`boolean_array_1` is a mask of `scores` because they have the same length.

`boolean_array_2` is not a mask of `scores` because they have different lengths.

## Selecting Elements of an Array Using a Mask

If `a` is a one-dimensional array and `m` is a mask of `a`, then `a[m]` contains the elements of `a` that correspond to `True` values in `m`.

````python
scores = np.array([82, 91, 74, 88, 65])
mask = np.array([True, True, False, True, False])
````

````python
scores[mask]
````

The values corresponding to `True` are kept, while the values corresponding to `False` are not.

## A Mask from a Condition

Thanks to broadcasting, comparing a one-dimensional numerical array with a single number produces a Boolean array of the same length.

Therefore, the result can be used as a mask.

````python
scores >= 80
````

The elements with indices 0, 1, and 3 are at least 80.

## Selecting the Elements That Satisfy a Condition

We can define a mask first:

````python
mask = scores >= 80
````

````python
scores[mask]
````

Since the condition itself creates the mask, we can also write the condition directly inside the brackets.

````python
scores[scores >= 80]
````

Both expressions return the same result.

## Element-by-Element `and` for Boolean Arrays

For NumPy Boolean arrays of the same length, we use `&` for element-by-element Boolean and.

````python
boolean_array_1 = np.array([True, True, False, True, False])
boolean_array_2 = np.array([False, True, True, False, False])
````

````python
boolean_array_1 & boolean_array_2
````

The result is `True` only where both arrays contain `True`.

## Combining Two Conditions with `&`

Suppose we want the scores that are at least 70 and less than 90.

Put each comparison inside parentheses.

````python
mask = (scores >= 70) & (scores < 90)
````

````python
scores[mask]
````

## Element-by-Element `or` for Boolean Arrays

For NumPy Boolean arrays of the same length, we use `|` for element-by-element Boolean or.

````python
boolean_array_1 | boolean_array_2
````

The result is `True` wherever at least one of the two arrays contains `True`.

## Combining Two Conditions with `|`

Suppose we want the scores that are less than 70 or greater than 90.

Again, put each comparison inside parentheses.

````python
mask = (scores < 70) | (scores > 90)
````

````python
scores[mask]
````

## The Main Pattern

The basic Boolean-indexing pattern is `array[condition]`.

The condition creates a Boolean mask, and NumPy keeps the elements corresponding to `True`.

## What You Should Remember

- A mask of a one-dimensional array is a Boolean array with the same length.
- `array[mask]` selects the elements corresponding to `True`.
- A comparison between an array and a number creates a Boolean mask.
- We can write `array[condition]` to select values satisfying a condition.
- Use `&` for element-by-element and.
- Use `|` for element-by-element or.
- Put each comparison inside parentheses when combining conditions.
