## From Python to Numerical Data

So far, we have worked with numbers, variables, lists, conditions, loops, and functions.

Now we begin learning tools designed specifically for numerical data.

## What Is NumPy?

NumPy is a Python library for working with numerical data.

Its name comes from **Numerical Python**.

NumPy helps us store collections of numbers and perform calculations on many numbers at once.

## What Is a Library?

A library is a collection of additional tools that programmers have created for us to use.

NumPy is one such library.

## Importing NumPy

Before using NumPy, we import it:

```python
import numpy as np
```

`np` is the commonly used abbreviation for NumPy.

````python
import numpy as np
````

It is normal that nothing is printed. We have simply made NumPy available in the notebook.

## We Already Know About Lists

A Python list can store several numbers:

```python
scores = [82, 91, 74]
```

NumPy provides another structure designed especially for numerical data: the **NumPy array**.

## A NumPy Array

We can create a NumPy array from a list using `np.array()`.

````python
scores = np.array([82, 91, 74])
````

````python
print(scores)
````

The values look similar to a Python list, but `scores` is now a NumPy array.

## Adding to a Python List

For an ordinary Python list, `scores + 5` does not mean add 5 to every element.

````python
scores = [82, 91, 74]
````

````python
# This produces an error:
# scores + 5
````

## Adding to a NumPy Array

With a NumPy array, the same idea works element by element.

````python
scores = np.array([82, 91, 74])
````

````python
scores + 5
````

## Vectorized Operations

Applying one operation to many values at once is called **vectorized computation**.

````python
scores + 5
````

````python
scores * 2
````

## Without NumPy, We Might Use a Loop

Using ordinary Python, we could process each value with a loop.

````python
scores = [82, 91, 74]

for score in scores:
    print(score + 5)
````

A NumPy array allows us to express the numerical operation more directly.

Both approaches are useful, but NumPy often makes numerical calculations shorter and easier to read.

## Why This Matters for Data

Real datasets may contain hundreds, thousands, or many more observations.

We often want to perform the same calculation on many values.

NumPy is designed for this kind of work.

## NumPy and Machine Learning

Machine-learning methods work extensively with numerical data.

Examples include ages, temperatures, house sizes, financial returns, and sensor measurements.

## NumPy Will Also Appear Inside Other Tools

Later in the course, we will use **Pandas** for tables of data.

Later, when you study machine learning, you will likely use libraries such as **scikit-learn**.

NumPy is part of the numerical foundation underneath much of the Python data-science ecosystem.

## What You Should Remember

- NumPy is a Python library designed for numerical data.
- Its central data structure is the NumPy array.
- NumPy allows us to perform calculations on many numerical values at once.
- `np.array()` creates a NumPy array.
- Vectorized computation applies one operation across many values.
