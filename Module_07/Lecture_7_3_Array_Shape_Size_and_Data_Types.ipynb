## Import NumPy

````python
import numpy as np
````

## Examining an Array

Once we have created an array, we often want to know:

- How many dimensions does it have?
- If it has two dimensions, how many rows and columns does it have?
- How many elements does it contain?
- What type of values does it contain?

NumPy provides this information directly.

## Useful Array Attributes

Four useful NumPy array attributes are:

```python
array.ndim
array.shape
array.size
array.dtype
```

An **attribute** stores information about an object.

We access an attribute by writing the array name, followed by a period and the attribute name.

Notice that we do **not** use parentheses after these attributes.

## A One-Dimensional Array

Create:

```python
scores = np.array([82, 91, 74])
```

````python
scores = np.array([82, 91, 74])
````

````python
scores
````

This array has one dimension and contains three elements.

## Number of Dimensions: `ndim`

The `ndim` attribute tells us the number of dimensions.

````python
scores.ndim
````

The result is:

```python
1
```

So `scores` is a one-dimensional array.

## Shape of a One-Dimensional Array

The `shape` attribute tells us the size of each dimension.

````python
scores.shape
````

The result is:

```python
(3,)
```

This means the array has one dimension containing 3 elements.

The comma is part of Python's notation for a tuple containing one element.

## Size of an Array

The `size` attribute counts the total number of elements.

````python
scores.size
````

For this array:

```python
shape = (3,)
size = 3
```

## A Two-Dimensional Array

Now create a two-dimensional array with two rows and three columns.

````python
data = np.array([[10, 20, 30],
                 [40, 50, 60]])
````

````python
data
````

## Dimensions and Shape

For a two-dimensional array:

```python
shape = (rows, columns)
```

````python
data.ndim
````

````python
data.shape
````

The results show:

- 2 dimensions,
- 2 rows,
- 3 columns.

## Size of a Two-Dimensional Array

The array has 2 rows and 3 columns, so it contains:

```text
2 × 3 = 6
```

elements.

````python
data.size
````

## Shape and Size Are Different

For this array:

```python
shape = (2, 3)
size = 6
```

`shape` tells us how the elements are organized.

`size` tells us the total number of elements.

## What Is `dtype`?

A NumPy array also has a data type.

We can examine it using:

```python
array.dtype
```

NumPy usually stores all elements in an array using one common data type.

## An Array of Integers

Create an array containing integers.

````python
numbers = np.array([10, 20, 30])
````

````python
numbers.dtype
````

NumPy reports an integer data type.

Depending on the computer and NumPy version, you may see a name such as `int64`.

You do not need to memorize the exact name now.

## An Array of Decimal Numbers

Now create an array containing decimal values.

````python
numbers = np.array([1.5, 2.5, 3.5])
````

````python
numbers.dtype
````

NumPy reports a floating-point data type.

You may see a name such as `float64`.

## NumPy Uses a Common Data Type

Consider:

```python
numbers = np.array([1, 2.5, 3])
```

Because one value is a floating-point number, NumPy chooses a floating-point type that can represent all the values.

````python
numbers = np.array([1, 2.5, 3])
````

````python
numbers
````

````python
numbers.dtype
````

Notice that the integers are stored and displayed as decimal values as well.

## Four Questions About an Array

When examining a NumPy array, we can ask:

- How many dimensions? → `array.ndim`
- What is its shape? → `array.shape`
- How many elements? → `array.size`
- What data type? → `array.dtype`

These attributes give us a quick description of an array before we begin working with its values.
