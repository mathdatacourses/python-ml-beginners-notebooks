## What Is Broadcasting?

Sometimes NumPy can combine an array and a single number, or arrays with different shapes.

NumPy may automatically reuse values so the operation can be performed.

This behavior is called **broadcasting**.

## We Have Already Used Broadcasting

Suppose `scores = np.array([82, 91, 74])`.

Then `scores + 5` adds 5 to every element.

This is the simplest example of broadcasting.

````python
scores = np.array([82, 91, 74])
````

````python
scores + 5
````

## Adding a Number to a Two-Dimensional Array

````python
data = np.array([[10, 20, 30],
                 [40, 50, 60]])
````

````python
data + 5
````

NumPy adds 5 to every element of `data`.

A single number can be combined with an array regardless of its shape.

## Broadcasting with a Two-Dimensional and a One-Dimensional Array

`data` has shape `(2, 3)` and `adjustment` has shape `(3,)`.

````python
data = np.array([[10, 20, 30],
                 [40, 50, 60]])

adjustment = np.array([1, 2, 3])
````

````python
data.shape
````

````python
adjustment.shape
````

## Adding the Arrays

If we calculate `data + adjustment`, NumPy adds the values 1, 2, and 3 to each row.

````python
data + adjustment
````

## What NumPy Is Doing

We can think of `adjustment = np.array([1, 2, 3])` as being reused for every row.

Conceptually, NumPy performs the operation on the first row and then the second row automatically.

## In Summary

Suppose `data` is a two-dimensional array and `adjustment` is a one-dimensional array.

If the length of `adjustment` equals the number of columns of `data`, then `data + adjustment` adds `adjustment` to each row of `data`.

We will not consider the case where the length of `adjustment` does not equal the number of columns of `data`.

## A Useful Interpretation

Suppose the three columns represent three different measurements.

For every row, we want to add 1 to column 0, 2 to column 1, and 3 to column 2.

Then `data + adjustment` performs all three adjustments for every row at once.

## Broadcasting Avoids Repetition

Broadcasting saves us from processing rows separately.

NumPy automatically applies the appropriate adjustment to every row.

## Rules to Remember

We will consider two broadcasting cases.

1. A single number can be combined with every element of an array, regardless of shape.
2. If a one-dimensional array has the same number of elements as the number of columns of a two-dimensional array, NumPy can apply that one-dimensional array to each row.

The same pattern works with subtraction, multiplication, and division.

## What You Should Remember

- Broadcasting allows NumPy to perform some operations between objects with different shapes.
- A single number can be combined with every element of an array.
- If a one-dimensional array has as many elements as a two-dimensional array has columns, NumPy can apply that array to each row.
- The same broadcasting pattern works with addition, subtraction, multiplication, and division.
