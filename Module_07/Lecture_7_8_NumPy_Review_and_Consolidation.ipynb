## Boolean Masks

```python
scores = np.array([82, 91, 74, 88, 65])
```

## A Condition Creates a Mask

```python
scores
```

## Combining Conditions

```python
scores
```

## Putting Several Ideas Together

Before running the code, predict what is stored in `adjusted`, `mask`, and `selected`.

```python
scores = np.array([72, 85, 91, 68, 88])
```

```python
scores
```

```python
adjusted = scores + 3
```

```python
adjusted
```

```python
mask = (adjusted >= 80) & (adjusted < 95)
```

```python
mask
```

```python
selected = adjusted[mask]
```

```python
selected
```
