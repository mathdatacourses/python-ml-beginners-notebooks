In this lecture, we learn how NumPy lets us perform arithmetic operations on entire arrays and on pairs of arrays without writing a loop for each element.

## Import NumPy

````python
import numpy as np
````

## Calculating with an Entire Array

One of the most useful features of NumPy is that we can perform calculations on many values at once.

Suppose:

```python
scores = np.array([82, 91, 74, 88])
```

If we write `scores + 5`, NumPy adds 5 to every element.

````python
scores = np.array([82, 91, 74, 88])
````

````python
scores + 5
````

## Other Arithmetic Operations

We can use the same idea with subtraction, multiplication, and division.

````python
scores - 10
````

````python
scores * 2
````

````python
scores / 2
````

## Division Produces Decimal Values

Division typically produces a floating-point array.

````python
numbers = np.array([10, 20, 30])
````

````python
numbers / 2
````

The decimal points indicate floating-point values.

## Powers

NumPy applies powers element by element.

````python
numbers = np.array([1, 2, 3, 4])
````

````python
numbers ** 2
````

## Vectorized Computation

Operations such as `scores + 5`, `scores * 2`, and `scores ** 2` apply an operation to an entire array at once.

This is called **vectorized computation**.

## Without Vectorization

Using a loop, we could write:

````python
scores = np.array([82, 91, 74, 88])

for score in scores:
    print(score + 5)
````

With NumPy, we can simply write `scores + 5`.

The NumPy expression is shorter and directly describes the numerical operation we want.

## Operations Between Two Arrays

NumPy can also perform arithmetic between arrays.

````python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3])
````

````python
a + b
````

NumPy adds elements that occupy corresponding positions.

## Element-by-Element Arithmetic

When two arrays have the same shape, NumPy can combine corresponding elements.

````python
a * b
````

## More Examples with Two Arrays

````python
a - b
````

````python
a / b
````

## The Original Array Does Not Change

Arithmetic does not automatically replace the original array.

````python
scores = np.array([82, 91, 74])
````

````python
scores + 5
````

````python
scores
````

## Saving the Result

If we want to keep the result, we assign it to a variable.

````python
new_scores = scores + 5
````

````python
scores
````

````python
new_scores
````

## Why Vectorized Arithmetic Matters

Vectorized arithmetic is useful for adjusting measurements, converting units, comparing sets of measurements, and transforming numerical variables.

## What You Should Remember

- NumPy arithmetic can operate on an entire array.
- A number can be combined with every element.
- Arrays of the same shape can be combined element by element.
- Addition, subtraction, multiplication, division, and powers all work this way.
- The original array does not change unless we assign a new result to it.
- This is called **vectorized computation**.
