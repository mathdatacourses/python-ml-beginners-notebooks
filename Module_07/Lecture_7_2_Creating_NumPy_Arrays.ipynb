## Creating an Array from a List

Suppose we already have a Python list:

```python
scores = [82, 91, 74]
```

We can convert it to a NumPy array using `np.array()`.

````python
scores = [82, 91, 74]
````

````python
scores = np.array(scores)
````

The values inside the array are still called **elements**.

## Displaying an Array

We can display the array by writing its variable name.

````python
scores
````

Colab displays something like:

```python
array([82, 91, 74])
```

We can also use `print()`.

````python
print(scores)
````

The display looks slightly different, but both are showing the same NumPy array.

## A One-Dimensional Array

The array:

```python
array([82, 91, 74])
```

is a **one-dimensional array**.

You can think of it as a single sequence of values.

Examples might include:

- test scores,
- temperatures,
- ages,
- financial returns.

## Creating a Two-Dimensional Array

NumPy can also organize numbers into rows and columns.

We create a two-dimensional array by placing each row in its own list, then placing those lists inside an outer list.

````python
data = np.array([[10, 20, 30],
                 [40, 50, 60]])
````

````python
data
````

````python
print(data)
````

## Rows and Columns

The array above has:

- 2 rows,
- 3 columns.

The first row contains 10, 20, and 30.

The second row contains 40, 50, and 60.

Two-dimensional arrays are useful for representing numerical information arranged in rows and columns.

## Creating an Array of Zeros

NumPy can create arrays for us.

For example:

```python
np.zeros(5)
```

creates an array containing five zeros.

````python
np.zeros(5)
````

The decimal points indicate the type of number NumPy created.

We will discuss NumPy data types in the next lecture.

## Creating an Array of Ones

Similarly:

```python
np.ones(4)
```

creates an array containing four ones.

````python
np.ones(4)
````

## Creating Sequences with `np.arange()`

NumPy provides `np.arange()`.

For example:

```python
np.arange(5)
```

creates:

```python
array([0, 1, 2, 3, 4])
```

As with `range()`, the stop value is not included.

````python
np.arange(5)
````

## `np.arange()` with Start and Stop

We can specify both a starting value and a stopping value.

The starting value is included.

The stopping value is not included.

````python
np.arange(2, 7)
````

## `np.arange()` with a Step

We can also specify a step.

The general form is:

```python
np.arange(start, stop, step)
```

````python
np.arange(2, 11, 2)
````

## `range()` and `np.arange()`

These look very similar:

```python
range(1, 6)
np.arange(1, 6)
```

Both represent the integer values 1 through 5.

But they are not the same type of object:

- `range()` is a built-in Python tool,
- `np.arange()` creates a NumPy array.

````python
np.arange(1, 6)
````

## Choose the Tool for the Task

If we mainly need values to control a Python `for` loop, `range()` is often convenient.

If we want to create a NumPy array for numerical calculations, `np.arange()` is often convenient.

We will use both.

- We can create NumPy arrays from Python lists using `np.array()`.
- NumPy arrays can be one-dimensional or two-dimensional.
- `np.zeros()` creates arrays of zeros.
- `np.ones()` creates arrays of ones.
- `np.arange()` creates numerical sequences as NumPy arrays.
