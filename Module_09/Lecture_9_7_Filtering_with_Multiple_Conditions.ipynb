We load the four-student dataset and set `Name` as the index of `df`.


```python
import pandas as pd

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/students_small.csv"
df = pd.read_csv(url)
df = df.set_index("Name")
```

## Our Starting DataFrame


```python
df
```

## Two Boolean Conditions

Suppose we want students who have `Score` at least 75 **and** are at least 17 years old.


```python
condition1 = df["Score"] >= 75
condition1
```

```python
condition2 = df["Age"] >= 17
condition2
```

Notice that the two Boolean Series have different patterns of `True` and `False` values.


## Combining Conditions with AND

In Pandas, use `&` for AND.

The combined result is `True` only when both original conditions are `True`.


```python
condition1 & condition2
```

## Filtering with the Combined Condition


```python
df[condition1 & condition2]
```

Only Ana satisfies both conditions.


## Writing AND in One Line

Put each comparison inside parentheses.


```python
df[(df["Score"] >= 75) & (df["Age"] >= 17)]
```

## Filtering with OR

Now suppose we want students who are 17 years old **or** have `Score` at least 90.

In Pandas, use `|` for OR.


```python
df[(df["Age"] == 17) | (df["Score"] >= 90)]
```

A row is kept if at least one of the two conditions is `True`.


## AND Versus OR

- Use `&` when **both** conditions must be `True`.
- Use `|` when **at least one** condition must be `True`.

When combining comparisons, put each comparison inside parentheses.


## Do Not Use `and` or `or`

For Pandas Boolean Series, use `&` for AND and `|` for OR.

Do not use Python's `and` or `or` to combine Series conditions.
