## Our Small DataFrame

We will continue using the four-student DataFrame `df`.


````python
df
````

## The Index of `df`

Every DataFrame has an index.

For this DataFrame, the default index labels are `0, 1, 2, 3`.


````python
df.index
````

Python displays:

```python
RangeIndex(start=0, stop=4, step=1)
```

This represents the labels `0, 1, 2, 3`.

The stopping value, `4`, is not included.


## Using Names as Row Labels

We can make the `Name` column the index and save the result in a new DataFrame called `df_by_name`.

The original DataFrame `df` remains unchanged.


````python
df_by_name = df.set_index("Name")
````

````python
df_by_name
````

### A note about unique index labels

In this example, every student has a different name.

In a real dataset, two students could have the same name. Pandas allows repeated index labels, so a unique student ID may sometimes be a better choice for the index.


## The New Index

The row labels are now the student names.


````python
df_by_name.index
````

## What Happened to the `Name` Column?

`Name` is no longer a regular column. It is now the index.


````python
df_by_name.columns
````

The index contains the row labels, while `columns` contains the column labels.


````python
df_by_name.index
````

````python
df_by_name.columns
````

## Selecting Maya with `loc`

`loc` selects rows by index label.


````python
df_by_name.loc["Maya"]
````

## Selecting Several Rows with `loc`

We can also use a slice with the name index.

With `loc`, the stopping label is included.


````python
df_by_name.loc["Ben":"Liam"]
````

This selects the rows labeled `Ben`, `Maya`, and `Liam`.


## Selecting the Third Row with `iloc`

`iloc` selects rows by position.

Even though the index labels are names, the rows still have positions `0, 1, 2, 3`.


````python
df_by_name.iloc[2]
````

## Same Row, Different Reason

These two expressions return Maya's row:

- `df_by_name.loc["Maya"]` uses the label `"Maya"`.
- `df_by_name.iloc[2]` uses position `2`.


````python
df_by_name.loc["Maya"]
````

````python
df_by_name.iloc[2]
````

## Which One Should We Use?

Use `loc` when you know the row label.

Use `iloc` when you know the row position.

For example, you may want the third row even if you do not know which student's name is there.

## What You Should Remember

- Every DataFrame has an index.
- `set_index()` can replace the default index with more meaningful labels.
- After making `Name` the index, `loc` selects students by name.
- `iloc` still selects rows by position.
- Use `loc` when you know the label.
- Use `iloc` when you know the position.
