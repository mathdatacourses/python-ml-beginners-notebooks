We load the four-student dataset and set `Name` as the index of `df`.


````python
import pandas as pd

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/students_small.csv"
df = pd.read_csv(url)
df = df.set_index("Name")
````

## Our Starting DataFrame


````python
df
````

## Creating a New Column

To create a new column, assign values to a new column name.

```python
df["NewColumn"] = values
```


## Example: Study Minutes

Convert `StudyHours` to minutes.


````python
df["StudyMinutes"] = df["StudyHours"] * 60
df
````

## Check the Columns

The new column is now part of `df`.


````python
df.columns
````

## Creating a Boolean Column

Create a column that is `True` when `Score` is at least 80.


````python
df["HighScore"] = df["Score"] >= 80
df
````

## More Examples of New Columns

The right side can be any expression that produces one value for each row.

```python
df["AgeNextYear"] = df["Age"] + 1
df["ScoreFraction"] = df["Score"] / 100
```


## New Name Versus Existing Name

- If the column name is new, Pandas creates a new column.
- If the column name already exists, Pandas changes that existing column.

Changing existing columns will be the topic of the next lecture.

## What You Should Remember

- Use `df["NewColumn"] = values` to create a new column.
- New values can come from expressions using existing columns.
- Pandas performs these operations row by row.
- Creating a new column modifies `df` by adding that column.
