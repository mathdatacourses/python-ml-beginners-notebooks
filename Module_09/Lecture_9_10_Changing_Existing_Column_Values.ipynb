## Setup

We load the four-student dataset and set `Name` as the index of `df`.


````python
import pandas as pd

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/students_small.csv"
df = pd.read_csv(url)
df = df.set_index("Name")
````

## Our Starting DataFrame


````python
df
````

## Changing an Existing Column

If we assign values to a column name that already exists, Pandas replaces the values in that column.

Suppose every student receives 5 additional points.


````python
df["Score"] = df["Score"] + 5
df
````

The `Score` column still exists, but its values have changed.

The number and names of the columns are unchanged.


````python
df.columns
````

## Changing One Value

Sometimes we want to change only one entry.

Suppose Maya's Score should be 70.

We can use `loc` to select Maya's row and the `Score` column, then assign the new value.


````python
df.loc["Maya", "Score"] = 70
df
````

## Selection and Assignment

Earlier, we used:

```python
df.loc["Maya", "Score"]
```

to select a value.

Now we can use:

```python
df.loc["Maya", "Score"] = 70
```

to replace that value.


## Important

Assignments change `df` itself.

If you want to return to the original data, rerun the setup cell.


## New Column or Existing Column?

If the column name is new:

```python
df["NewColumn"] = values
```

Pandas creates a new column.

If the column name already exists:

```python
df["Score"] = values
```

Pandas replaces the values in that column.

## What You Should Remember

- Assigning to an existing column changes that column.
- Example: `df["Score"] = df["Score"] + 5`
- To change one specific value, use `df.loc[row_label, column_label] = new_value`.
- You can also use `iloc` with row and column positions.
- These assignments modify `df` itself.
