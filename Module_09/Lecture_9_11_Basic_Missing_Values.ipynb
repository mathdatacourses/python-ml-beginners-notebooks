## Setup

Run this cell before starting the demonstrations and practice.

We load the four-student dataset and set `Name` as the index of `df`.


```python
import pandas as pd

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/students_small.csv"
df = pd.read_csv(url)
df = df.set_index("Name")
```

## Our Starting DataFrame


```python
df
```

## Creating Missing Values

Our original dataset has no missing values.

For this lecture, we will intentionally make Ben's and Maya's `StudyHours` values missing.


```python
df.loc["Ben", "StudyHours"] = pd.NA
df.loc["Maya", "StudyHours"] = pd.NA
df
```

## Finding Missing Values

Use `isna()` to check every entry in the DataFrame.

The result is `True` when a value is missing and `False` otherwise.


```python
df.isna()
```

## Counting Missing Values

Use `isna().sum()` to count the missing values in each column.


```python
df.isna().sum()
```

## Option 1: Remove Rows with Missing Values

`dropna()` returns a DataFrame with rows containing missing values removed.


```python
df.dropna()
```

Notice that `dropna()` returns a new DataFrame.

It does not automatically change `df`.


```python
df
```

## Option 2: Fill Missing Values

`fillna()` can replace missing values.

Here we replace missing `StudyHours` values with 0 only to demonstrate how the method works.


```python
df["StudyHours"].fillna(0)
```

At this point, `df` itself has not changed.

To save the filled values in `df`, assign the result back to the column.


```python
df["StudyHours"] = df["StudyHours"].fillna(0)
df
```

## Important

Replacing a missing value with 0 is not automatically the correct choice.

A missing `StudyHours` value does not necessarily mean the student studied 0 hours.

The appropriate treatment depends on what the missing values mean.

## What You Should Remember

- Missing numeric values are often displayed as `NaN`.
- Use `isna()` to locate missing values.
- Use `isna().sum()` to count them.
- `dropna()` can remove rows containing missing values.
- `fillna()` can replace missing values.
- The appropriate choice depends on the meaning of the data.
