## Selecting a Row with `iloc`

`iloc` selects rows by **position**.

The first row is at position `0`.


````python
df.iloc[0]
````

The third row is at position `2`.


````python
df.iloc[2]
````

## Selecting a Row with `loc`

`loc` selects rows by **index label**.

With the default index, the labels are `0, 1, 2, 3`.


````python
df.loc[0]
````

````python
df.loc[3]
````

## Comparing `iloc` and `loc`

With the default integer index, `df.iloc[0]` and `df.loc[0]` happen to return the same row.

They do so for different reasons:

- `iloc[0]` means position `0`.
- `loc[0]` means index label `0`.

In the next lecture, we will use labels that are names rather than numbers so the difference is easier to see.


````python
df.iloc[0]
````

````python
df.loc[0]
````

## Selecting Several Rows with `iloc`

Python-style slicing excludes the stopping position.


````python
df.iloc[0:3]
````

`df.iloc[0:3]` selects positions `0`, `1`, and `2`.

Position `3` is not included.


## Selecting Several Rows with `loc`

With `loc`, the stopping **label is included**.


````python
df.loc[0:3]
````

`df.loc[0:3]` selects the rows with labels `0`, `1`, `2`, and `3`.


## Compare the Two Slices

Compare:

```python
df.iloc[0:3]
```

with:

```python
df.loc[0:3]
```

The first returns 3 rows, while the second returns 4 rows.


````python
df.iloc[0:3]
````

````python
df.loc[0:3]
````

## Single Row Versus Several Rows

Selecting one row with `iloc` or `loc` returns a Series.

Selecting a range of rows with a slice returns a DataFrame.


````python
type(df.iloc[0])
````

````python
type(df.iloc[0:3])
````

## What You Should Remember

- `iloc` selects rows by **position**.
- `loc` selects rows by **index label**.
- `iloc` slices exclude the stop position.
- `loc` slices include the stop label.
- Selecting one row returns a Series.
- Selecting a range of rows returns a DataFrame.
