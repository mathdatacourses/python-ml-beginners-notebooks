## Our Starting DataFrame


```python
df
```

## One Argument or Two?

With one argument, `loc` and `iloc` select rows.

With two arguments, the first selects rows and the second selects columns.


```python
df.loc["Maya"]
```

## Maya's Score with `loc`


```python
df.loc["Maya", "Score"]
```

## The Same Value with `iloc`


```python
df.iloc[2, 1]
```

## One Row and Several Columns


```python
df.loc["Maya", ["Age", "Score"]]
```

## Several Rows and Several Columns


```python
df.loc["Ben":"Liam", ["Age", "Score"]]
```

The same rows can also be listed explicitly.


```python
df.loc[["Ben", "Maya", "Liam"], ["Age", "Score"]]
```

## The Same Block with `iloc`


```python
df.iloc[1:4, 0:2]
```

## What You Should Remember

- With one argument, `loc` and `iloc` select rows.
- With two arguments, the first selects rows and the second selects columns.
- `loc` uses labels.
- `iloc` uses positions.
- Rows come first, and columns come second.
