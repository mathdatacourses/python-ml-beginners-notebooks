## Setup

Run this cell before starting the demonstrations and practice.


```python
import pandas as pd

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/students_small.csv"
df = pd.read_csv(url)
df = df.set_index("Name")
```

## Our Starting DataFrame


```python
df
```

## Sorting by One Column

By default, `sort_values()` sorts in ascending order.


```python
df.sort_values("Score")
```

## Sorting in Descending Order


```python
df.sort_values("Score", ascending=False)
```

## The Original DataFrame Is Unchanged


```python
df
```

Save the sorted result in another variable if you want to keep it.


```python
df_sorted = df.sort_values("Score")
df_sorted
```

## Sorting by Another Column


```python
df.sort_values("Age")
```

Ana and Maya are both 17 years old. A second column can break ties.


## Sorting by More Than One Column


```python
df.sort_values(["Age", "Score"])
```

## Different Directions for Different Columns


```python
df.sort_values(["Age", "Score"], ascending=[True, False])
```

## What You Should Remember

- Use `sort_values()` to order rows by column values.
- The default order is ascending.
- Use `ascending=False` for descending order.
- You can sort by one column or several columns.
- Later columns break ties in earlier columns.
- `sort_values()` does not change the original `df` unless you save the sorted result.
