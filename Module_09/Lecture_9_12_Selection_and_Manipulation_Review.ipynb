## Putting Several Skills Together

Suppose we want to:

1. keep students with `Score` at least 75,
2. sort them by `Score` from largest to smallest,
3. display only `Score` and `StudyHours`.

We can solve this one step at a time.


```python
selected = df[df["Score"] >= 75]
selected
```

```python
selected = selected.sort_values("Score", ascending=False)
selected
```

```python
selected[["Score", "StudyHours"]]
```

Each line performs one familiar operation.

Together, the lines solve a larger task.
