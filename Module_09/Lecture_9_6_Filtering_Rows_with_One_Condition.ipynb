## Setup

Run this cell before starting the demonstrations and practice.


```python
import pandas as pd

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/students_small.csv"
df = pd.read_csv(url)
df = df.set_index("Name")
```

## Our Starting DataFrame


```python
df
```

## Step 1: Create a Condition

Suppose we want students with `Score` at least 80.


```python
df["Score"] >= 80
```

The result is a Boolean Series with one `True` or `False` value for each row.


## Step 2: Use the Condition to Filter


```python
condition = df["Score"] >= 80
condition
```

```python
df[condition]
```

## Write the Filter in One Line


```python
df[df["Score"] >= 80]
```

## Filtering with Equality


```python
df[df["Age"] == 17]
```

## Another Example


```python
df[df["StudyHours"] < 5]
```

## Filtering a Boolean Column


```python
df[df["Passed"] == False]
```

## What You Should Remember

- A condition produces one `True` or `False` value for each row.
- Use `df[condition]` to keep rows where the condition is `True`.
- You can save a condition in a variable or write it directly inside `df[...]`.
- Comparison operators such as `>`, `>=`, `<`, `<=`, `==`, and `!=` can create conditions.
