## Selecting a Single Column

To select one column, write the column name inside square brackets.

```python
df["Score"]
```

The result is a Pandas Series.


````python
df["Score"]
````

## Selecting Other Columns

The same pattern works for any column.


````python
df["Name"]
````

````python
df["Age"]
````

````python
df["Passed"]
````

## Checking the Type

Selecting one column produces a Pandas Series.


````python
type(df["Score"])
````

## The Column Name Is a String

In `df["Score"]`, `"Score"` is a Python string, so the column name is written between quotation marks.


## Column Names Are Case-Sensitive

`df["Score"]` and `df["score"]` are not the same. The column name must match the DataFrame exactly.
