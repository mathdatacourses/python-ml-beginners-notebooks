## Selecting Two Columns

Select the `Name` and `Score` columns.


```python
df[["Name", "Score"]]
```

## One Column: Series Versus DataFrame

Compare `df["Score"]` with `df[["Score"]]`.


```python
df["Score"]
```

```python
df[["Score"]]
```

## Checking the Types

Use `type()` to compare the two results.


```python
type(df["Score"])
```

```python
type(df[["Score"]])
```

## Selecting More Than Two Columns


```python
df[["Name", "Age", "Score"]]
```

## Selecting Columns That Are Not Next to Each Other


```python
df[["Name", "StudyHours", "Passed"]]
```

## The Order Comes from the List


```python
df[["Score", "Name"]]
```
