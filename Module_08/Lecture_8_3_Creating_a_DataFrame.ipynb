## Import Pandas

````python
import pandas as pd
````

## Creating Our First DataFrame

A DataFrame organizes data into:

- rows,
- columns,
- column names,
- a row index.

We will create a small DataFrame containing information about students.

## Start with the Data

Suppose we have:

| Name | Age | Score | Passed |
|---|---:|---:|---|
| Ana | 17 | 88 | True |
| Ben | 18 | 74 | True |
| Maya | 17 | 63 | False |

We want to store this table in a Pandas DataFrame.

## A Python Dictionary Can Organize the Columns

One convenient way to begin is with a Python dictionary.

````python
data = {
    "Name": ["Ana", "Ben", "Maya"],
    "Age": [17, 18, 17],
    "Score": [88, 74, 63],
    "Passed": [True, True, False]
}
````

Each dictionary key is the name of a column.

## The Dictionary Keys Become Column Names

The keys:

```text
Name, Age, Score, Passed
```

will become the DataFrame column names.

## The Lists Become Column Values

Each key is associated with a list.

For example:

```python
"Score": [88, 74, 63]
```

means that the `Score` column contains 88, 74, and 63.

The values in the same positions across the lists belong to the same row.

## Creating the DataFrame

We create the DataFrame using:

```python
df = pd.DataFrame(data)
```

````python
df = pd.DataFrame(data)
````

The variable `df` now refers to a Pandas DataFrame.

## Displaying the DataFrame

````python
df
````

The column names are `Name`, `Age`, `Score`, and `Passed`.

The numbers on the left are the row index.

## The Default Row Index

Because we did not provide row labels ourselves, Pandas created the default index:

```text
0, 1, 2
```

For now, we will usually use this default index.

## Each List Must Have the Same Length

When creating a DataFrame from a dictionary of lists, the lists must have the same length.

In our example, each list contains three values.

This allows Pandas to form three complete rows.

## What If the Lengths Are Different?

Suppose one list contains three values and another contains only two.

Pandas cannot form complete rows from these lists.

Trying to create the DataFrame produces an error.

````python
bad_data = {
    "Name": ["Ana", "Ben", "Maya"],
    "Score": [88, 74]
}

# Try removing the # from the next line to see the error:
# bad_df = pd.DataFrame(bad_data)
````

## Checking the Type

We can use `type()` to confirm that `df` is a Pandas DataFrame.

````python
type(df)
````

## A DataFrame Can Contain Different Types

Our DataFrame contains:

- `Name` — text,
- `Age` — integers,
- `Score` — integers,
- `Passed` — Boolean values.

Different columns can store different kinds of information.

## A Useful Pattern

A common way to create a small DataFrame is:

```python
data = {
    "Column1": [...],
    "Column2": [...],
    "Column3": [...]
}

df = pd.DataFrame(data)
```

The dictionary organizes the columns.

`pd.DataFrame()` creates the table.
