## What Is a CSV File?

CSV stands for **Comma-Separated Values**.

A CSV file is a simple text file used to store data in a table.

Each line usually represents one row, and commas separate the values into columns.

## A Simple CSV File

A CSV file might contain:

```text
Name,Age,Score
Ana,17,88
Ben,18,74
Maya,17,63
```

The first line usually contains the column names.

This first line is often called the **header**.

The remaining lines contain the data.

## The CSV File Represents a Table

The CSV text above represents this table:

| Name | Age | Score |
|---|---:|---:|
| Ana | 17 | 88 |
| Ben | 18 | 74 |
| Maya | 17 | 63 |

The CSV file stores the same information using plain text.

## Why Are CSV Files Useful?

CSV files are commonly used because they are:

- simple,
- easy to share,
- supported by many programs,
- convenient for storing tables of data.

Many datasets used in data science and machine learning are stored as CSV files.

## Pandas Can Read CSV Files

Pandas provides:

```python
pd.read_csv()
```

to read data stored in a CSV file.

The result is a Pandas DataFrame.

## A CSV File Can Be Available on the Internet

Sometimes a CSV file is publicly available on the internet.

A file on the internet can have a **URL**.

A URL is the web address that tells us where the file is located.

If the CSV file is public, Pandas can read it directly from its URL.

## Reading a CSV File from a URL

First import Pandas.

````python
import pandas as pd
````

Store the web address of the CSV file in a variable called `url`.

````python
url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/students.csv"
````

Now load the CSV file into a DataFrame.

````python
df = pd.read_csv(url)
````

Pandas accesses the CSV file and creates a DataFrame.

No manual upload is needed.

## The Result Is a DataFrame

After:

```python
df = pd.read_csv(url)
```

the variable `df` refers to a Pandas DataFrame.

## Displaying the DataFrame

````python
df
````

Colab displays the data as a table.

The values from the CSV header become the DataFrame column names.

Pandas also creates the default row index.

## Looking at the First Few Rows

Large datasets may contain hundreds or thousands of rows.

Instead of displaying the entire DataFrame, we can use:

```python
df.head()
```

By default, this displays the first 5 rows.

````python
df.head()
````

Our example DataFrame is small, so `head()` may display all of its rows.

## CSV Files Can Come from Different Places

A CSV file might be:

- publicly available on the internet,
- stored on your computer,
- stored in your Google Drive.

For now, we will use public CSV files from the internet.

## A Useful Pattern

A common pattern is:

```python
import pandas as pd

df = pd.read_csv(url)

df.head()
```

The first line imports Pandas.

The second loads the dataset.

The third lets us inspect the first few rows.

## What You Should Remember

- A CSV file stores tabular data in a text file.
- CSV stands for **Comma-Separated Values**.
- Pandas can load a CSV file using `pd.read_csv()`.
- A public CSV file can be read directly from its URL.
- The result is a Pandas DataFrame.
- The CSV header becomes the DataFrame column names.
- We can inspect the first few rows using `df.head()`.
