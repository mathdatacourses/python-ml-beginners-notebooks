## Import Pandas

```python
import pandas as pd
```

## Two Important Pandas Objects

Pandas provides two important objects for organizing data: **Series** and **DataFrame**.

A Series is one-dimensional.

A DataFrame is two-dimensional and organizes data into rows and columns.

## What Is a Series?

A Pandas Series is a one-dimensional labeled sequence of values.

The labels are called the **index**.

## Creating a Simple Series

```python
scores = pd.Series([88, 74, 63])
```

## Displaying a Series

```python
scores
```

The numbers on the left are the index labels.

The numbers on the right are the values.

## The Default Index

When we do not provide index labels, Pandas creates them automatically.

For this Series, the default index is `0, 1, 2`.

## Series and NumPy Arrays

A one-dimensional NumPy array and a Pandas Series both store a one-dimensional sequence of values.

A Series also has index labels.

## Checking the Type

```python
type(scores)
```

## From a Series to a DataFrame

A Series is useful for one sequence of values.

A dataset usually contains several variables, so Pandas also provides the DataFrame.

## What Is a DataFrame?

A Pandas DataFrame is a two-dimensional labeled table.

| Name | Age | Score | Passed |
|---|---:|---:|---|
| Ana | 17 | 88 | True |
| Ben | 18 | 74 | True |
| Maya | 17 | 63 | False |

## A DataFrame Has Two Kinds of Labels

A DataFrame has column labels and row labels.

For the student table, the column labels are `Name`, `Age`, `Score`, and `Passed`.

With the default row index, the row labels are `0, 1, 2`.

## DataFrame Columns Are Series

Each column of a DataFrame is a Pandas Series.

The Series uses the same row index as the DataFrame.

## Different Columns Can Have Different Types

`Name` contains text, `Age` and `Score` contain integers, and `Passed` contains Boolean values.

## Series or DataFrame?

- A Series is one-dimensional.
- A DataFrame is two-dimensional.

A Series often represents one variable, while a DataFrame organizes several variables together.
