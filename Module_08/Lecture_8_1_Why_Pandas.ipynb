## From NumPy Arrays to Tables of Data

In Module 7, we learned how to work with numerical data using NumPy arrays.

Real datasets are often organized as tables with rows representing observations, columns representing variables, names for the columns, and different types of information.

Pandas provides tools designed for working with this kind of tabular data.

## A Typical Dataset

| Name | Age | Score | Passed |
|---|---:|---:|---|
| Ana | 17 | 88 | True |
| Ben | 18 | 74 | True |
| Maya | 17 | 63 | False |

This table contains 3 rows of observations, 4 variables, and numerical, text, and Boolean information.

## Why Not Use Only NumPy?

NumPy is excellent for numerical arrays.

But real datasets may contain numerical columns, text columns, Boolean columns, and meaningful column names.

For tabular datasets like these, Pandas provides a convenient way to organize and work with the data.

## What Is Pandas?

Pandas is a Python library for working with tabular data.

It provides tools for organizing data into rows and columns, giving columns meaningful names, loading data from files, selecting and filtering data, and sorting and modifying data.

## Importing Pandas

Before using Pandas, we import it:


## The DataFrame

One of the most important objects in Pandas is called a **DataFrame**.

A DataFrame organizes data into a table.

It has rows, columns, and column names.

## Rows Represent Observations

Each row represents one **observation**.

In this example, each observation is one student.

## Columns Represent Variables

The columns represent variables.

In this example, the variables are `Name`, `Age`, `Score`, and `Passed`.

## Column Names Matter

In a NumPy array, we often work with numerical positions.

In a DataFrame, columns can have meaningful names such as `Name`, `Age`, and `Score`.

Meaningful column names make it easier to understand what the data represents.

## Different Columns Can Store Different Types

A DataFrame can contain different types of information in different columns.

For example, `Name` contains text, `Age` and `Score` contain integers, and `Passed` contains Boolean values.

## Pandas and NumPy Work Together

Pandas does not replace NumPy.

The two libraries are often used together.

NumPy is especially useful for numerical arrays and numerical calculations, while Pandas is especially useful for labeled tables of data.
