## Why Pandas?

Pandas is a Python library designed for working with tabular data.

Tabular data is organized into rows and columns.

## Series and DataFrames

A Pandas Series is a one-dimensional collection of values.

A Pandas DataFrame is a two-dimensional table with rows, columns, column names, and a row index.

## Creating a DataFrame

````python
import pandas as pd
````

````python
data = {
    "Name": ["Ana", "Ben", "Maya"],
    "Age": [17, 18, 17],
    "Score": [88, 74, 63]
}
````

````python
df = pd.DataFrame(data)
````

````python
df
````

When we create a DataFrame from a dictionary of lists, the dictionary keys become the column names and the lists contain the column values.

The lists must have the same length.

## CSV Files

CSV stands for **Comma-Separated Values**.

A CSV file stores tabular data using plain text.

The first line often contains the column names and is called the header.

## Loading a CSV File

````python
url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/students.csv"
````

````python
df = pd.read_csv(url)
````

The CSV file is now loaded into a Pandas DataFrame.

## Inspecting the First and Last Rows

````python
df.head()
````

````python
df.tail()
````

````python
df.head(3)
````

````python
df.tail(2)
````

The number inside the parentheses controls how many rows are displayed.

## Inspecting the Structure

````python
df.shape
````

````python
df.columns
````

````python
df.index
````

````python
df.size
````

`df.shape` gives the number of rows and columns.

`df.columns` gives the column names.

`df.index` gives information about the row index.

`df.size` gives the total number of values.

## Inspecting Data Types

````python
df.dtypes
````

Common data types include `int64`, `float64`, `str`, and `bool`.

In some versions of Pandas, text columns may appear as `object`.

## Getting a Broader Summary

````python
df.info()
````

`df.info()` gives a broader summary that includes information about the rows and columns, non-null counts, and column data types.

## Methods and Attributes

Methods use parentheses:

```python
df.head()
df.tail()
df.info()
```

Attributes do not:

```python
df.shape
df.columns
df.index
df.dtypes
```
