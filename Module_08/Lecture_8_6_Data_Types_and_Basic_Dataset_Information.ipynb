## Load the Student Dataset

```python
import pandas as pd
```

```python
url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/students.csv"
```

```python
df = pd.read_csv(url)
```

## Finding the Data Types

Use `df.dtypes` to see the data type of each column.

```python
df.dtypes
```

The student DataFrame contains text, integer, floating-point, and Boolean columns.

Depending on the Pandas version, the text column may appear as `str` or `object`.

## Integer Columns

`int64` is used for integer values such as those in `Age` and `Score`.

## Floating-Point Columns

`float64` is used for floating-point values such as those in `StudyHours`.

## Boolean Columns

`bool` is used for Boolean values such as `True` and `False` in `Passed`.

## Text Columns

Text columns can appear as `str`. In some Pandas versions, they may instead appear as `object`.

## Why Do Data Types Matter?

Data types affect what operations make sense for a column. We can perform arithmetic with numerical columns, use string operations with text, and interpret Boolean columns as `True` or `False` conditions.

## Getting More Information About a DataFrame

Use `df.info()` for a compact summary of the DataFrame.

```python
df.info()
```

## Information Shown by `info()`

For each column, `df.info()` shows the column name, non-null count, and data type.

It also reports overall DataFrame information such as the number of rows.

## What Does Non-Null Count Mean?

The Non-Null Count tells us how many values are present in a column.

For example, `Age    12 non-null` means that 12 values are present in the `Age` column.

## `dtypes` Versus `info()`

`df.dtypes` focuses on the data type of each column.

`df.info()` gives a broader summary that includes rows and columns, non-null counts, and data types.

## An Attribute and a Method

`df.dtypes` is an attribute, so it does not use parentheses.

`df.info()` is a method, so it uses parentheses.
