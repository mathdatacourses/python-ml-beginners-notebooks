## Looking at the First Rows

```python
df.head()
```

## Looking at the Last Rows

```python
df.tail()
```

```python
df.tail(3)
```

Looking at both the beginning and the end can help us understand what the data looks like and check that the file loaded correctly.

## The Shape of a DataFrame

```python
df.shape
```

The result has the form `(number of rows, number of columns)`. The order is always **rows first, then columns**.

## Finding the Column Names

```python
df.columns
```

## Looking at the Row Index

```python
df.index
```

The default Pandas index begins at zero. For 12 rows, the index labels are 0 through 11.

## The Size of a DataFrame

```python
df.size
```

For a DataFrame with 12 rows and 3 columns, `12 * 3 = 36`, so `df.size` is 36.

## `shape` Versus `size`

- `df.shape` gives the number of rows and columns separately.
- `df.size` gives the total number of values.

## Methods and Attributes

`df.head()` and `df.tail()` use parentheses.

`df.shape`, `df.columns`, `df.index`, and `df.size` do not.

`head()` and `tail()` are **methods**. Methods perform an action and are called using parentheses.

`shape`, `columns`, `index`, and `size` are **attributes**. Attributes store information about an object.

## Comparing a Method with an Attribute

```python
df.head
```

```python
df.head()
```

```python
df.shape
```

## A Useful Inspection Pattern

After loading a dataset, we might use `df.head()`, `df.tail()`, `df.shape`, `df.columns`, and `df.index` to get a quick first look.

## What You Should Remember

- `df.head()` inspects the first rows.
- `df.tail()` inspects the last rows.
- `df.head(n)` and `df.tail(n)` let us choose how many rows to display.
- `df.shape` gives the number of rows and columns.
- `df.columns` gives the column names.
- `df.index` gives information about the row index.
- `df.size` gives the total number of values.
- `head()` and `tail()` are methods and use parentheses.
- `shape`, `columns`, `index`, and `size` are attributes and do not use parentheses.
