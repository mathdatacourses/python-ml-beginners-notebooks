## Values Have Types

Python works with different kinds of values.

Examples include:

```python
5
3.14
"hello"
True
```

Python calls these different kinds of values **data types**.

## Four Basic Python Data Types

| Python Type | Meaning | Example |
|---|---|---|
| `int` | Integer | `5` |
| `float` | Decimal number | `3.14` |
| `str` | Text | `"hello"` |
| `bool` | True or False | `True` |

## Integers: `int`

An integer is a whole number.

Let's ask Python for the type of two integer values.

````python
type(5)
````

````python
type(-12)
````

## Decimal Numbers: `float`

Numbers containing a decimal point are usually stored as floating-point numbers.

Even `5.0` is a float.

````python
type(5)
````

````python
type(5.0)
````

## Text: `str`

Text is stored using a Python type called a **string**.

Strings are placed inside quotation marks.

````python
type("hello")
````

````python
name = "Anna"
````

````python
type(name)
````

## Numbers Inside Quotes Are Text

Compare:

```python
82
```

with:

```python
"82"
```

Without quotation marks, Python interprets `82` as a number.

With quotation marks, Python interprets `"82"` as text.

````python
type(82)
````

````python
type("82")
````

## Boolean Values: `bool`

A Boolean value is either `True` or `False`.

Notice that `True` and `False` begin with capital letters.

````python
type(True)
````

````python
type(False)
````

## Variables Can Store Different Types

The variable name does not determine the type.

The value assigned to the variable determines its type.

````python
age = 16
````

````python
temperature = 72.5
````

````python
name = "Anna"
````

````python
passed = True
````

````python
type(age)
````

````python
type(temperature)
````

````python
type(name)
````

````python
type(passed)
````

## Finding a Value's Type

The `type()` function asks Python what type a value has.

For example:

```python
type(5)
type(3.14)
```

We will study functions more carefully later in the course.

## Python Data Types and Dataset Variable Types

Earlier in the course, we discussed dataset variables such as:

- numerical variables,
- categorical variables.

Python data types such as `int`, `float`, `str`, and `bool` are related, but they are not the same idea.

Dataset variable types describe what a variable means in a dataset.

Python data types describe how individual values are represented in Python.

## What You Should Remember

- `int` represents whole numbers.
- `float` represents decimal numbers.
- `str` represents text.
- `bool` represents `True` or `False`.
- Quotation marks tell Python that a value should be treated as text.
- Use `type()` to ask Python for the data type of a value.
- Python data types are not the same thing as numerical and categorical dataset variable types.
