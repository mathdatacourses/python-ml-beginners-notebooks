## Python Can Evaluate Arithmetic Expressions

Python can evaluate arithmetic expressions.

An expression is a combination of values and operations that Python can evaluate.

```python
2 + 3
```

```python
10 + 7
```

## Basic Arithmetic Operators

Python uses operators for arithmetic:

- `+` for addition
- `-` for subtraction
- `*` for multiplication
- `/` for division
- `**` for powers

## Addition, Subtraction, and Multiplication

Remember that Python uses an asterisk `*` for multiplication.

```python
8 + 5
```

```python
10 - 3
```

```python
6 * 4
```

## Division

Python uses the forward slash `/` for division.

```python
10 / 2
```

```python
7 / 2
```

Notice that Python displays `5.0` for `10 / 2`.

We will discuss why when we introduce Python data types.

## Powers

Python uses two asterisks `**` to represent a power.

One asterisk means multiplication.

Two asterisks represent a power.

```python
2 ** 3
```

```python
5 ** 2
```

## Order of Operations

Python follows the usual order of arithmetic operations:

1. Parentheses
2. Powers
3. Multiplication and division
4. Addition and subtraction

Parentheses can be used to change the order.

```python
2 + 3 * 4
```

```python
(2 + 3) * 4
```

## A Few Important Python Symbols

Some familiar mathematical notation is written differently in Python.

| Meaning | Write in Python |
|---|---|
| Multiply | `3 * 4` |
| Divide | `10 / 2` |
| Square | `5 ** 2` |

Programming languages require precise syntax.
