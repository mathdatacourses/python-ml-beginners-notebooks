## Why Do We Need Variables?

A variable gives a value a name so that we can refer to it again later.

## Creating a Variable

Create a variable called `x` and assign it the value `5`.

````python
x = 5
````

Running an assignment statement does not normally display an output.

Now enter the variable name by itself.

````python
x
````

## The Assignment Operator

In Python, the symbol `=` is called the **assignment operator**.

It assigns the value on the right to the variable name on the left.

````python
score = 82
````

````python
score
````

## Using Variables in Expressions

Once a variable has a value, we can use it in calculations.

````python
hours = 5
````

````python
hours * 10
````

## Using More Than One Variable

Descriptive variable names can make calculations easier to understand.

````python
price = 12
````

````python
quantity = 4
````

````python
price * quantity
````

## A Variable Can Be Reassigned

A variable can later be assigned a new value.

The new assignment replaces the previous value associated with that variable.

````python
x = 5
````

````python
x
````

````python
x = 10
````

````python
x
````

## Why This Looks Strange at First

In Python, the statement

```python
x = x + 1
```

is not a mathematical equation.

It means:

1. take the current value of `x`,
2. add 1,
3. assign the new result back to `x`.

````python
x = 5
````

````python
x = x + 1
````

````python
x
````

## Choosing Variable Names

Good variable names describe what the values represent.

Because Python variable names cannot contain spaces, words are commonly separated with underscores.

````python
exam_score = 82
````

````python
exam_score
````

## Basic Variable-Name Rules

Variable names:

- cannot contain spaces,
- may contain letters, digits, and underscores,
- cannot begin with a digit,
- are case-sensitive.

A common Python style is to use lowercase words separated by underscores, such as:

```python
hours_studied
```
