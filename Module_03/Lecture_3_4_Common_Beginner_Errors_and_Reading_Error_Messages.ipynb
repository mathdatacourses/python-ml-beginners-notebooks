## Errors Are a Normal Part of Programming

When Python cannot carry out an instruction, it displays an error message.

Errors happen to everyone who writes code.

A useful first goal is not to avoid every error, but to learn how to use the information Python gives us to find and correct the problem.


Programming languages require precise instructions.

## `TypeError`

Sometimes Python understands the instruction but cannot perform the requested operation with the types of values provided.

First, add two numbers.

````python
5 + 3
````

Now try to add an integer to a string.

The next cell is **supposed to produce a `TypeError`**.

````python
5 + "3"
````

The integer `5` and the string `"3"` are different types.

Python cannot add them in this way.

## Start With the Last Line

Python error messages can sometimes look long.

At this stage, you do not need to understand every line.

A good first step is:

**Look at the last line of the error message.**

It often tells you:

- the type of error,
- a short description of what went wrong.

## Look Back at the Code

After reading the last line, ask simple questions:

- Did I spell the variable name correctly?
- Did I include all quotation marks and parentheses?
- Did I use a valid variable name?
- Am I combining values of compatible types?

Many beginner errors can be fixed by checking these details carefully.

## Do Not Be Afraid to Experiment

When Python gives you an error:

1. Read the last line.
2. Look carefully at the code.
3. Make a small correction.
4. Run the cell again.

Programming often involves repeating this process.
