## Strings Represent Text

Recall that Python uses the type `str` to represent text.

Examples include:

```python
"hello"
"Anna"
"Python is useful"
```

A string is a sequence of characters.

Those characters can include letters, numbers, spaces, and punctuation.

## Single or Double Quotation Marks

Python accepts both double quotation marks and single quotation marks when creating strings.

For consistency, we will usually use double quotation marks in this course.

````python
type("hello")
````

````python
type('hello')
````

## Strings Can Be Stored in Variables

We can assign a string to a variable just as we assign numbers.

````python
name = "Anna"
````

````python
name
````

````python
type(name)
````

## Combining Strings

We can combine strings using the `+` operator.

Joining strings together in this way is called **concatenation**.

````python
"Hello " + "Anna"
````

````python
name = "Anna"
````

````python
"Hello " + name
````

## Spaces Matter

Python does not automatically insert a space between strings.

A space is itself a character.

````python
"Hello" + "Anna"
````

````python
"Hello " + "Anna"
````

## Repeating a String

We can repeat a string using multiplication.

````python
"Hi" * 3
````

````python
"-" * 10
````

## Finding the Length of a String

Python provides the function `len()` to count the number of characters in a string.

Spaces also count as characters.

````python
len("Python")
````

````python
len("Data Science")
````

## Strings Are Sequences of Characters

Consider the string:

```python
"Python"
```

It contains six characters:

| Character | P | y | t | h | o | n |
|---|---|---|---|---|---|---|
| Index | 0 | 1 | 2 | 3 | 4 | 5 |

Python begins counting positions at **0**, not 1.

These positions are called **indices**.

## Selecting One Character

We can select one character using square brackets.

````python
word = "Python"
````

````python
word[0]
````

````python
word[2]
````

## Spaces Also Have Indices

Spaces are characters too.

In the string `"Data Science"`, the space is at index 4.

````python
text = "Data Science"
````

````python
text[4]
````

## Selecting Part of a String

We can select several characters from a string using a **slice**.

The starting index is included.

The ending index is not included.

````python
word = "Python"
````

````python
word[0:3]
````

A useful rule to remember is:

**Start included, end excluded.**

````python
word[2:6]
````

- Strings represent text in Python.

- We can repeat strings using `*`.
- We can find the number of characters using `len()`.
- Strings are sequences of characters.
