## Python Can Perform Operations on Strings

Python provides useful operations for working with text.

For example, we may want to:

- convert letters to lowercase,
- convert letters to uppercase,
- remove extra spaces,
- replace part of a string,
- check whether text appears inside another string.

## What Is a Method?

A method is a function associated with a particular object.

For example:

```python
"Hello".lower()
```

The dot connects the string to the method:

```python
string.method()
```

We will see this dot notation many times later in Python.

## Converting to Lowercase

The method `.lower()` converts letters to lowercase.

````python
"Hello".lower()
````

````python
"PYTHON".lower()
````

## Converting to Uppercase

The method `.upper()` converts letters to uppercase.

````python
"Python".upper()
````

````python
"Data Science".upper()
````

## Removing Extra Spaces

The method `.strip()` removes spaces from the beginning and end of a string.

Spaces inside the text are not removed.

````python
" Anna ".strip()
````

````python
" Data Science ".strip()
````

## Why `strip()` Is Useful

Two strings can look similar but still be different if one contains extra spaces.

````python
city1 = "Atlanta"
city2 = " Atlanta "
````

````python
city1 == city2
````

````python
city1 == city2.strip()
````

## Replacing Text

The method `.replace()` replaces one part of a string with another.

The first argument tells Python what to look for.

The second tells Python what to use instead.

````python
"Data Science".replace("Data", "Computer")
````

````python
"Data Science".replace("Science", "Analysis")
````

## Checking Whether Text Is Present

The operator `in` checks whether one string appears inside another.

The result is a Boolean value.

````python
"Science" in "Data Science"
````

````python
"Python" in "Data Science"
````

## Methods Can Be Used with Variables

String methods also work on strings stored in variables.

````python
name = "Anna"
````

````python
name.lower()
````

````python
name.upper()
````

## The Original String Does Not Automatically Change

String methods usually return a new string.

They do not automatically replace the original value stored in a variable.

````python
name = "Anna"
````

````python
name.lower()
````

````python
name
````

If we want to save the lowercase version, we assign the result back to the variable.

````python
name = name.lower()
````

````python
name
````

## Combining String Methods

Methods can sometimes be used one after another.

This is called **method chaining**.

Python performs the operations from left to right.

````python
name = " Anna "
````

````python
name.strip().lower()
````

## What You Should Remember

- `.lower()` converts letters to lowercase.
- `.upper()` converts letters to uppercase.
- `.strip()` removes spaces from the beginning and end of a string.
- `.replace()` replaces part of a string with different text.
- `in` checks whether one string appears inside another.
- String methods usually return a new string.
- They do not automatically change the original string.
- To save the new result, assign it back to a variable.
- Methods can be chained together when several operations are needed in sequence.
