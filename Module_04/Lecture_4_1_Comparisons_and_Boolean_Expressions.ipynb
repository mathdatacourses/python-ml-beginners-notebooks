In this lecture, we learn how Python compares values and produces Boolean results: `True` or `False`.


## Python Can Compare Values

A comparison is an expression that Python can evaluate.

For example, Python can ask whether one number is greater than another.


````python
5 > 3
````

````python
2 > 10
````

## Comparison Operators

Python provides several comparison operators.

| Operator | Meaning | Example |
|---|---|---|
| `==` | Equal to | `5 == 5` |
| `!=` | Not equal to | `5 != 3` |
| `>` | Greater than | `7 > 4` |
| `<` | Less than | `2 < 8` |
| `>=` | Greater than or equal to | `5 >= 5` |
| `<=` | Less than or equal to | `4 <= 9` |


## Comparisons Produce Boolean Values

Every comparison produces one of two Boolean values:

```python
True
```

or

```python
False
```


````python
8 < 12
````

````python
10 == 7
````

## Equal To: `==`

To ask whether two values are equal, Python uses **two equals signs**.


````python
5 == 5
````

````python
5 == 8
````

## `=` and `==` Are Different

This distinction is very important.

One equals sign is used for **assignment**:

```python
x = 5
```

Two equals signs are used for **comparison**:

```python
x == 5
```


````python
x = 5
````

````python
x == 5
````

````python
x == 8
````

## Not Equal To: `!=`

The operator `!=` means **not equal to**.


````python
5 != 3
````

````python
5 != 5
````

## Greater Than and Less Than

Python uses the familiar greater-than and less-than symbols.

It also provides `>=` and `<=` when equality should count.


````python
8 > 5
````

````python
3 < 1
````

````python
5 >= 5
````

````python
4 <= 9
````

## Comparing Variables

We can compare variables just as we compare numbers directly.


````python
score = 82
````

````python
score >= 70
````

````python
score > 90
````

## A Comparison Is a Boolean Expression

Recall that an expression is something Python can evaluate.

A comparison is also an expression.

For example:

```python
age >= 18
```

Python evaluates the comparison and produces either `True` or `False`.

An expression that produces a Boolean value is called a **Boolean expression**.

## What You Should Remember

- Python comparison operators include `==`, `!=`, `>`, `<`, `>=`, and `<=`.
- Comparisons produce Boolean values: `True` or `False`.
- A comparison is a Boolean expression.
- One equals sign, `=`, means assignment.
- Two equals signs, `==`, mean comparison.
