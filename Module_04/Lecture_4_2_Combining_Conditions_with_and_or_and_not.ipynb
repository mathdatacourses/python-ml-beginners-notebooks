In this lecture, we learn how to combine Boolean expressions using the logical operators `and`, `or`, and `not`.

## Combining Boolean Expressions

In the previous lecture, we created Boolean expressions such as:

```python
score >= 70
```

and:

```python
age < 18
```

Sometimes we want to combine several conditions.

Python provides three important logical operators:

```python
and
or
not
```

## The `and` Operator

The operator `and` requires **both conditions to be `True`**.

For example:

```python
(age >= 18) and (age <= 65)
```

is `True` only when both comparisons are `True`.

````python
age = 25
````

````python
(age >= 18) and (age <= 65)
````

````python
age = 70
````

````python
(age >= 18) and (age <= 65)
````

## `and`: Both Must Be True

| Condition 1 | Condition 2 | Condition 1 `and` Condition 2 |
|---|---|---|
| `True` | `True` | `True` |
| `True` | `False` | `False` |
| `False` | `True` | `False` |
| `False` | `False` | `False` |

The key idea is:

**`and` means both conditions must be `True`.**

## The `or` Operator

The operator `or` requires **at least one condition to be `True`**.

For example:

```python
(score >= 90) or (attendance >= 95)
```

is `True` if either comparison is `True`, or if both are `True`.

````python
score = 85
attendance = 97
````

````python
(score >= 90) or (attendance >= 95)
````

````python
score = 85
attendance = 90
````

````python
(score >= 90) or (attendance >= 95)
````

## `or`: At Least One Must Be True

| Condition 1 | Condition 2 | Condition 1 `or` Condition 2 |
|---|---|---|
| `True` | `True` | `True` |
| `True` | `False` | `True` |
| `False` | `True` | `True` |
| `False` | `False` | `False` |

The key idea is:

**`or` means at least one condition must be `True`.**

## The `not` Operator

The operator `not` works with one Boolean value.

It reverses the value.

```python
not True
```

becomes `False`, and:

```python
not False
```

becomes `True`.

````python
not True
````

````python
not False
````

## Using `not` with a Variable

Python can also reverse the Boolean value stored in a variable.

````python
passed = True
````

````python
not passed
````

````python
passed = False
````

````python
not passed
````

## A Practical Example

Suppose a student qualifies only if:

- the exam score is at least 70, and
- attendance is at least 80.

We can write:

```python
(score >= 70) and (attendance >= 80)
```

````python
score = 75
attendance = 85
````

````python
(score >= 70) and (attendance >= 80)
````

````python
attendance = 70
````

````python
(score >= 70) and (attendance >= 80)
````

## Use Parentheses for Clarity

When combining conditions, parentheses can make the code easier to read.

For example:

```python
(age >= 18) and (age <= 65)
```

and:

```python
(score >= 90) or (attendance >= 95)
```

The parentheses are not always required in simple expressions like these, but they make the individual comparisons easier to see.
