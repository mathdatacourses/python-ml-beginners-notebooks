## Setup

Import Pandas and Matplotlib, then load the club dataset into a DataFrame called `df`.


````python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/club_students.csv"
df = pd.read_csv(url)
````

## The DataFrame


````python
df
````

## Create a Bar Chart

Use the `Club` column for the categories on the horizontal axis and the `Students` column for the heights of the bars.


````python
plt.bar(df["Club"], df["Students"])
plt.show()
````

## The Order of the Arguments

In

```python
plt.bar(df["Club"], df["Students"])
```

the first argument provides the categories on the horizontal axis.

The second argument provides the heights of the bars.


## Make the Bar Chart Clear and Readable

Use the same formatting ideas from the previous lectures.


````python
plt.figure(figsize=(4, 3))

plt.bar(df["Club"], df["Students"])

plt.xlabel("Club", fontsize=12)
plt.ylabel("Students", fontsize=12)

plt.title("Students in Each Club", fontsize=14)

plt.show()
````

## Change the Bar Color

The `color` parameter can be used to choose a color for the bars.

If we do not specify a color, Matplotlib chooses one automatically.


````python
plt.figure(figsize=(4, 3))

plt.bar(df["Club"], df["Students"], color="green")

plt.xlabel("Club", fontsize=12)
plt.ylabel("Students", fontsize=12)

plt.title("Students in Each Club", fontsize=14)

plt.show()
````
