In this lecture, we create scatter plots directly from Pandas DataFrame columns.


## Setup

Import Pandas and Matplotlib, then load the four-student dataset into `df`.


```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/students_small.csv"
df = pd.read_csv(url)
```

## Our DataFrame


```python
df
```

## Select the Two Columns

Each selected DataFrame column is a Pandas Series.


```python
df["StudyHours"]
```

```python
df["Score"]
```

## Save the Series in Variables

We use standard Python `snake_case` names for variables.


```python
study_hours = df["StudyHours"]
score = df["Score"]
```

## Create the Scatter Plot from the Series

`study_hours` is the first argument, so it provides the horizontal coordinates.

`score` is the second argument, so it provides the vertical coordinates.


```python
plt.scatter(study_hours, score)
plt.show()
```

## Use the DataFrame Columns Directly

We do not need to create separate variables first.


```python
plt.scatter(df["StudyHours"], df["Score"])
plt.show()
```

The underlying values are the same, so the scatter plot is the same.

## What You Should Remember

- Matplotlib can use Pandas Series directly.
- The first argument to `plt.scatter()` gives the horizontal coordinates.
- The second argument gives the vertical coordinates.
- You can pass DataFrame columns directly to Matplotlib.
- There is no need to convert Pandas columns into lists.
