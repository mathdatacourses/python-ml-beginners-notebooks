## Setup for the Instructor Example

Import Pandas and Matplotlib, then load the student dataset into a DataFrame called `df`.

```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/student_scores_studyhours.csv"
df = pd.read_csv(url)

df.head()
```

## From a Question to a Plot

Question:

**Does Score appear to be related to StudyHours?**

Both variables are numerical, and we want to examine their relationship.

A scatter plot is an appropriate choice.

## Putting the Workflow Together

Create, format, save, and display the scatter plot.

```python
plt.figure(figsize=(4, 3))

plt.scatter(df["StudyHours"], df["Score"])

plt.xlabel("Study Hours", fontsize=12)
plt.ylabel("Score", fontsize=12)

plt.title("Score versus Study Hours", fontsize=14)

plt.savefig("score_vs_study_hours.png", dpi=300)

plt.show()
```

## Review: Choosing the Right Plot

- **Scatter plot** — relationship between two numerical variables
- **Line plot** — change across an ordered variable
- **Bar chart** — compare numerical values across categories
- **Histogram** — distribution of one numerical variable

## What You Should Remember

- Use a scatter plot for relationships between two numerical variables.
- Use a line plot for change across an ordered variable.
- Use a bar chart to compare values across categories.
- Use a histogram to examine the distribution of one numerical variable.
- Use labels and titles to make figures clear.
- Use `plt.legend()` when multiple labeled lines appear on the same figure.
- Use `plt.savefig()` when you want to save a figure.
