# Lecture 10.4 — Making Scatter Plots Clear and Readable

This notebook accompanies Lecture 10.4 of **Python, NumPy, Pandas & Matplotlib for Machine Learning Beginners**.

In this lecture, we improve a scatter plot by changing the figure size and point size, and by adding axis labels, a title, and readable font sizes.


## Setup

Import Pandas and Matplotlib, then load the same four-student dataset used in Lecture 10.3.


```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/students_small.csv"
df = pd.read_csv(url)
```

## Our DataFrame


```python
df
```

## Starting with Our Scatter Plot

This is the basic scatter plot from Lecture 10.3.


```python
plt.scatter(df["StudyHours"], df["Score"])
plt.show()
```

## Change the Figure Size

Use `plt.figure(figsize=(4, 3))` before creating the plot.

The first number is the width of the figure in inches, and the second number is the height.


```python
plt.figure(figsize=(4, 3))

plt.scatter(df["StudyHours"], df["Score"])
plt.show()
```

## Change the Size of the Points

For a scatter plot, the parameter `s` controls the size of the points.

Here we use `s=80`.


```python
plt.figure(figsize=(4, 3))

plt.scatter(df["StudyHours"], df["Score"], s=80)
plt.show()
```

## Add Axis Labels

Use `plt.xlabel()` to label the horizontal axis and `plt.ylabel()` to label the vertical axis.


```python
plt.figure(figsize=(4, 3))

plt.scatter(df["StudyHours"], df["Score"], s=80)

plt.xlabel("Study Hours")
plt.ylabel("Score")

plt.show()
```

## Add a Title

Use `plt.title()` to add a title to the figure.


```python
plt.figure(figsize=(4, 3))

plt.scatter(df["StudyHours"], df["Score"], s=80)

plt.xlabel("Study Hours")
plt.ylabel("Score")

plt.title("Score versus Study Hours")

plt.show()
```

## Change the Font Sizes

We use font size 12 for the axis labels and font size 14 for the title.


```python
plt.figure(figsize=(4, 3))

plt.scatter(df["StudyHours"], df["Score"], s=80)

plt.xlabel("Study Hours", fontsize=12)
plt.ylabel("Score", fontsize=12)

plt.title("Score versus Study Hours", fontsize=14)

plt.show()
```

## Putting Everything Together

The complete code below creates a clear and readable scatter plot.


```python
plt.figure(figsize=(4, 3))

plt.scatter(df["StudyHours"], df["Score"], s=80)

plt.xlabel("Study Hours", fontsize=12)
plt.ylabel("Score", fontsize=12)

plt.title("Score versus Study Hours", fontsize=14)

plt.show()
```
