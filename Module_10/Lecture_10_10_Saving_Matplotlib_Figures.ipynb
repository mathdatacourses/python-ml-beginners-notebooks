# Lecture 10.10 — Saving Matplotlib Figures

This notebook accompanies Lecture 10.10 of **Python, NumPy, Pandas & Matplotlib for Machine Learning Beginners**.

In this lecture, we will save Matplotlib figures as files.


## Setup

Import Pandas and Matplotlib, then load the same student dataset used in Lecture 10.9 into a DataFrame called `df`.


````python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/student_scores_studyhours.csv"
df = pd.read_csv(url)

df.head(3)
````

## Create the Score Histogram

This is the same histogram used in the previous lecture.


````python
plt.figure(figsize=(4, 3))

plt.hist(df["Score"], bins=5, range=(50, 100))

plt.xlabel("Score", fontsize=12)
plt.ylabel("Number of Students", fontsize=12)

plt.title("Distribution of Scores", fontsize=14)

plt.show()
````

## Save the Figure

Use `plt.savefig()` before `plt.show()`.

The filename is written as a string.


````python
plt.figure(figsize=(4, 3))

plt.hist(df["Score"], bins=5, range=(50, 100))

plt.xlabel("Score", fontsize=12)
plt.ylabel("Number of Students", fontsize=12)

plt.title("Distribution of Scores", fontsize=14)

plt.savefig("score_histogram.png")

plt.show()
````

## Where Is the File Saved in Colab?

If you provide only a filename, Colab normally saves the file in the current working directory:

`/content`

The file should appear in the Files panel on the left.

Files in the Colab runtime are temporary, so download any file you want to keep.


````python
# Optional: check the current working directory
import os
os.getcwd()
````

## Kaggle

In a Kaggle notebook, files created by the notebook are normally stored in:

`/kaggle/working`

For example:

`/kaggle/working/score_histogram.png`

Files in `/kaggle/working` can be preserved as notebook outputs when you save a notebook version.


## Save a Higher-Resolution Image

Use the `dpi` parameter to control the resolution of the saved PNG.


````python
plt.figure(figsize=(4, 3))

plt.hist(df["Score"], bins=5, range=(50, 100))

plt.xlabel("Score", fontsize=12)
plt.ylabel("Number of Students", fontsize=12)

plt.title("Distribution of Scores", fontsize=14)

plt.savefig("score_histogram.png", dpi=300)

plt.show()
````

## Other File Types

The filename extension determines the file type.

For example:

```python
plt.savefig("figure.png")
plt.savefig("figure.pdf")
```

For this course, we will usually save figures as PNG files.
