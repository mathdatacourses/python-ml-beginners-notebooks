# Lecture 10.6 — Customizing Lines and Markers

This notebook accompanies Lecture 10.6 of **Python, NumPy, Pandas & Matplotlib for Machine Learning Beginners**.

In this lecture, we will customize the appearance of a line plot using line width, line style, markers, marker size, and color.


## Setup

Import Pandas and Matplotlib, then load the Day and Temperature dataset into a DataFrame called `df`.


```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/temperature_days.csv"
df = pd.read_csv(url)
```

## The DataFrame


```python
df
```

## Starting with Our Line Plot

By default, Matplotlib displays a solid line and does not mark the individual data points.


```python
plt.plot(df["Day"], df["Temperature"])
plt.show()
```

## Change the Line Width

The parameter `linewidth` controls the thickness of the line.

A larger value produces a thicker line.


```python
plt.plot(df["Day"], df["Temperature"], linewidth=3)
plt.show()
```

## Change the Line Style

The parameter `linestyle` controls how the line is drawn.

The value `"--"` creates a dashed line.


```python
plt.plot(df["Day"], df["Temperature"], linestyle="--")
plt.show()
```

## Add Markers

The parameter `marker` displays a symbol at each data point.

The value `"o"` uses circles as markers.


```python
plt.plot(df["Day"], df["Temperature"], marker="o")
plt.show()
```

## Change the Marker Size

The parameter `markersize` controls the size of the markers.


```python
plt.plot(
    df["Day"],
    df["Temperature"],
    marker="o",
    markersize=8
)

plt.show()
```

## Change the Color

The parameter `color` controls the color of the line and markers.

Here we use the simple color name `"green"`.


```python
plt.plot(
    df["Day"],
    df["Temperature"],
    marker="o",
    markersize=8,
    color="green"
)

plt.show()
```

## Putting Everything Together

Now combine the customization options with the figure size, axis labels, and title from the previous lecture.


```python
plt.figure(figsize=(4, 3))

plt.plot(
    df["Day"],
    df["Temperature"],
    linewidth=2,
    linestyle="--",
    marker="o",
    markersize=8,
    color="green"
)

plt.xlabel("Day", fontsize=12)
plt.ylabel("Temperature", fontsize=12)

plt.title("Temperature versus Day", fontsize=14)

plt.show()
```

## What You Should Remember

- `linewidth` controls line thickness.
- `linestyle` controls how the line is drawn.
- `marker` displays symbols at the data points.
- `markersize` controls marker size.
- `color` controls the color of the line and markers.
- These parameters change the appearance of the graph, not the underlying data.
