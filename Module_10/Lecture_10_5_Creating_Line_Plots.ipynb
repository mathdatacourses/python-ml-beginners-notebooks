In this lecture, we will create line plots directly from columns of a Pandas DataFrame.


## Setup

Import Pandas and Matplotlib, then load the Day and Temperature dataset used in the lecture into a DataFrame called `df`.


````python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/temperature_days.csv"
df = pd.read_csv(url)
````

## The DataFrame


````python
df
````

## Create a Line Plot

Use the `Day` column for the horizontal coordinates and the `Temperature` column for the vertical coordinates.


````python
plt.plot(df["Day"], df["Temperature"])
plt.show()
````

## The Order of the Arguments

In

```python
plt.plot(df["Day"], df["Temperature"])
```

the first argument provides the horizontal coordinates, and the second argument provides the vertical coordinates.

So this is a plot of **Temperature versus Day**.


## Make the Line Plot Clear and Readable

Use the formatting ideas from Lecture 10.4.


````python
plt.figure(figsize=(4, 3))

plt.plot(df["Day"], df["Temperature"])

plt.xlabel("Day", fontsize=12)
plt.ylabel("Temperature", fontsize=12)

plt.title("Temperature versus Day", fontsize=14)

plt.show()
````

## What You Should Remember

- Use `plt.plot(x, y)` to create a line plot.
