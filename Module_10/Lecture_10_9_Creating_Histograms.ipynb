In this lecture, we will use histograms to show how the values of one numerical variable are distributed.


## Setup

Import Pandas and Matplotlib, then load the student dataset into a DataFrame called `df`.

The dataset contains both `Score` and `StudyHours`.


```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/student_scores_studyhours.csv"
df = pd.read_csv(url)
```

## The DataFrame


```python
df
```

## Create a Histogram

A histogram needs only one numerical column.

Here, we create a histogram of `Score`.


```python
plt.hist(df["Score"])
plt.show()
```

## Control the Number of Bins

Use the `bins` parameter to choose the number of intervals.

Here we use 5 bins.


```python
plt.hist(df["Score"], bins=5)
plt.show()
```

## Control the Histogram Range

By default, Matplotlib uses the smallest and largest data values to determine the histogram range.

Use the `range` parameter to choose where the leftmost bin starts and the rightmost bin ends.

With 5 bins and a range from 50 to 100, the bin boundaries are:

**50, 60, 70, 80, 90, 100**


```python
plt.hist(df["Score"], bins=5, range=(50, 100))
plt.show()
```

## Bin Boundary Convention

For these bins, Matplotlib places a boundary value such as `60` in the bin that starts at `60`.

So a score of 60 belongs to the 60–70 bin.


## Make the Histogram Clear and Readable

Now add the figure size, axis labels, and title.


```python
plt.figure(figsize=(4, 3))

plt.hist(df["Score"], bins=5, range=(50, 100))

plt.xlabel("Score", fontsize=12)
plt.ylabel("Number of Students", fontsize=12)

plt.title("Distribution of Scores", fontsize=14)

plt.show()
```

## Interpreting the Histogram

- The horizontal axis shows intervals of `Score` values.
- The vertical axis shows how many students have scores in each interval.
- The histogram helps us see where the scores are concentrated.

## What You Should Remember

- A histogram shows the distribution of one numerical variable.
- Use `plt.hist(df["column"])` to create a histogram.
- Use `bins=...` to control the number of bins.
- Use `range=(lower, upper)` to control the histogram range.
- A histogram differs from a bar chart because its horizontal axis represents numerical intervals rather than categories.
