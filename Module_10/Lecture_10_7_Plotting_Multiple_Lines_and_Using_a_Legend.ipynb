# Lecture 10.7 — Plotting Multiple Lines and Using a Legend

This notebook accompanies Lecture 10.7 of **Python, NumPy, Pandas & Matplotlib for Machine Learning Beginners**.

In this lecture, we will plot more than one line on the same figure and use a legend to identify the lines.


## Setup

Import Pandas and Matplotlib, then load the High and Low temperature dataset into a DataFrame called `df`.


```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/mathdatacourses/python-ml-beginners-data/main/high_low_temperatures.csv"
df = pd.read_csv(url)
```

## The DataFrame


```python
df
```

## Plot the First Line

Plot `High` versus `Day`.


```python
plt.plot(df["Day"], df["High"])
plt.show()
```

## Add a Second Line

Call `plt.plot()` again before `plt.show()`.

By default, Matplotlib uses a different color for each new line.


```python
plt.plot(df["Day"], df["High"])
plt.plot(df["Day"], df["Low"])

plt.show()
```

## Add Labels to the Lines

Use the `label` parameter to give each line a name.


```python
plt.plot(df["Day"], df["High"], label="High")
plt.plot(df["Day"], df["Low"], label="Low")

plt.show()
```

## Display the Legend

The labels do not appear automatically.

Use `plt.legend()` to display them.


```python
plt.plot(df["Day"], df["High"], label="High")
plt.plot(df["Day"], df["Low"], label="Low")

plt.legend()

plt.show()
```

## Complete Figure

Now add the figure size, axis labels, title, and legend.

We also use `fontsize=10` for the legend text.


```python
plt.figure(figsize=(4, 3))

plt.plot(df["Day"], df["High"], label="High")
plt.plot(df["Day"], df["Low"], label="Low")

plt.xlabel("Day", fontsize=12)
plt.ylabel("Temperature", fontsize=12)

plt.title("High and Low Temperature", fontsize=14)

plt.legend(fontsize=10)

plt.show()
```

## What to Notice

Each call to `plt.plot()` adds another line to the same figure.

The two lines use different colors automatically.

The colors help distinguish the lines, while the legend identifies what each line represents.
