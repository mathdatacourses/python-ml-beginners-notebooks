## Programs Can Make Decisions

Sometimes we want Python to perform an action only when a condition is `True`.

For example, we might want to display `"Passed"` if a score is at least 70, display a warning if a temperature is too high, or perform a calculation only when data is available.

Python uses an `if` statement to make this kind of decision.

## A Simple `if` Statement

Suppose:

```python
score = 82
```

We can write:

```python
if score >= 70:
    print("Passed")
```

Because the condition is `True`, Python displays `Passed`.

````python
score = 82

if score >= 70:
    print("Passed")
````

## The Condition Is a Boolean Expression

The expression `score >= 70` is a Boolean expression.

It evaluates to either `True` or `False`.

The indented code runs only when the condition is `True`.

## Using `print()`

The function `print()` displays information.

````python
print("Hello")
````

````python
score = 82
print(score)
````

## What If the Condition Is False?

If the condition is `False`, Python skips the indented code.

Nothing is printed, and that is not an error.

````python
score = 60

if score >= 70:
    print("Passed")
````

## The Colon Is Required

An `if` statement ends its condition with a colon.

Leaving out the colon causes a syntax error.

## Indentation Is Required

The code controlled by the `if` statement must be indented.

Indentation is part of Python syntax.

````python
score = 82

if score >= 70:
    print("Passed")
````

````python
# This would produce an indentation error:
# if score >= 70:
# print("Passed")
````

## Several Instructions Can Be Controlled

An `if` statement can control more than one instruction.

All indented lines belong to the same block.

````python
temperature = 95

if temperature > 90:
    print("Hot day")
    print("Drink water")
````

## Indentation Defines the Block

Only indented lines depend on the condition.

A line that returns to the left margin is outside the `if` block.

````python
score = 60

if score >= 70:
    print("Passed")

print("Finished")
````

## Conditions Can Use Logical Operators

We can use `and`, `or`, and `not` inside an `if` condition.

The example below prints `Qualified` only if both conditions are `True`.

````python
score = 80
attendance = 85

if (score >= 70) and (attendance >= 80):
    print("Qualified")
````

````python
attendance = 75

if (score >= 70) and (attendance >= 80):
    print("Qualified")
````

## A Simple Pattern

The basic structure is:

```python
if condition:
    instructions
```

Python evaluates the condition, runs the indented code if the condition is `True`, and skips it if the condition is `False`.

## What You Should Remember

- An `if` statement allows Python to make a decision.
- The condition must evaluate to `True` or `False`.
- If the condition is `True`, Python runs the indented block.
- If the condition is `False`, Python skips the indented block.
- Put a colon after the condition.
- Indent the code controlled by the `if` statement.
- Indentation determines which instructions belong to the block.
