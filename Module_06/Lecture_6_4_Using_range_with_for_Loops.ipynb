## Repeating a Certain Number of Times

In the previous lecture, we used a `for` loop to move through the elements of a list.

Sometimes we simply want to repeat an instruction a certain number of times, or work with a sequence of integers.

Python provides the function:

```python
range()
```

## Our First Example with `range()`

Consider:

```python
for number in range(5):
    print(number)
```

Python prints the integers from 0 through 4.

````python
for number in range(5):
    print(number)
````

## What Does `range(5)` Mean?

`range(5)` represents:

```python
0, 1, 2, 3, 4
```

Two important facts:

- it starts at 0,
- it stops before 5.

So `range(5)` contains five integers, but 5 itself is not included.

## Following the Loop Step by Step

In:

```python
for number in range(5):
    print(number)
```

the loop variable `number` takes the values:

```python
0, 1, 2, 3, 4
```

one at a time.

For each value, Python runs:

```python
print(number)
```

After 4, the loop ends.

## The Stop Value Is Not Included

This is an important rule:

```python
range(stop)
```

starts at 0 and stops before `stop`.

For example:

```python
range(3)   -> 0, 1, 2
range(5)   -> 0, 1, 2, 3, 4
range(10)  -> 0, 1, ..., 9
```

````python
for number in range(3):
    print(number)
````

````python
for number in range(5):
    print(number)
````

````python
for number in range(10):
    print(number)
````

## Starting Somewhere Other Than Zero

We can provide both a starting value and a stopping value.

The general form is:

```python
range(start, stop)
```

The starting value is included.

The stopping value is not included.

````python
for number in range(1, 6):
    print(number)
````

## Start Is Included, Stop Is Not

Consider:

```python
range(3, 8)
```

This represents:

```python
3, 4, 5, 6, 7
```

So 3 is included and 8 is not.

A useful phrase to remember is:

**Start here, stop before there.**

## Changing the Step

By default, `range()` increases by 1.

We can choose a different step:

```python
range(start, stop, step)
```

For example:

```python
range(2, 11, 2)
```

represents:

```python
2, 4, 6, 8, 10
```

## Using a Step in a Loop

````python
for number in range(2, 11, 2):
    print(number)
````

Now change the step from 2 to 3 and compare the result.

````python
for number in range(2, 11, 3):
    print(number)
````

## Repeating an Action

Sometimes we use `range()` only because we want an action to happen a certain number of times.

We do not always need to use the value of the loop variable.

````python
for number in range(3):
    print("Practice Python")
````

````python
for number in range(5):
    print("Practice Python")
````

## Three Common Forms of `range()`

The three forms are:

```python
range(stop)
range(start, stop)
range(start, stop, step)
```

Examples:

```python
range(5)
range(1, 6)
range(2, 11, 2)
```

Remember: the stop value is never included.

## What You Should Remember

- `range()` is commonly used with `for` loops to generate a sequence of integers.
- `range(stop)` starts at 0.
- `range(start, stop)` lets us choose the starting value.
- `range(start, stop, step)` also lets us choose the step.
- The start value is included.
- The stop value is not included.
- `range()` can also control how many times an action repeats.
