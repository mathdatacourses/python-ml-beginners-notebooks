## Why Do We Need Functions?

Suppose we have several instructions that we want to use more than once.

We could copy and paste the same code every time.

But this can make programs longer, harder to read, and harder to change.

A function allows us to give a name to a block of code and use it whenever we need it.

## Defining Our First Function

```python
def say_hello():
    print("Hello")
```

The word `def` tells Python that we are defining a function.

````python
def say_hello():
    print("Hello")
````

Defining the function does not run it.

Python stores the instructions under the function's name.

## The Structure of a Function

The basic structure is:

```python
def function_name():
    instructions
```

Notice that `def` begins the definition, parentheses follow the name, the line ends with a colon, and the function body is indented.

## Defining Is Not the Same as Running

To run the function, we must call it.

````python
say_hello()
````

## Calling a Function More Than Once

Once a function is defined, we can call it whenever we need it.

````python
say_hello()
say_hello()
say_hello()
````

## Functions Can Receive Information

Functions become more useful when they can work with different values.

We can do this using a **parameter**.

## A Function with a Parameter

```python
def greet(name):
    print("Hello", name)
```

Here, `name` is a parameter.

````python
def greet(name):
    print("Hello", name)
````

## Calling the Function with an Argument

Now call the function with a value.

````python
greet("Ana")
````

The value `"Ana"` is an **argument**.

- A **parameter** is the variable in the function definition.
- An **argument** is the value supplied when the function is called.

## Different Arguments, Different Results

````python
greet("Ana")
greet("David")
greet("Maya")
````

## A Function Can Return a Value

Functions can calculate a result and send that result back.

````python
def square(number):
    return number ** 2
````

````python
square(5)
````

## What Does `return` Do?

`return` sends a value back to the place where the function was called.

````python
result = square(5)
print(result)
````

## `print()` and `return` Are Different

`print()` displays a value on the screen.

`return` sends a value back from the function.

````python
def show_square(number):
    print(number ** 2)

def square(number):
    return number ** 2
````

## Using a Returned Value

Because `square()` returns a value, we can store it and use it in another calculation.

````python
result = square(5)
answer = result + 10
print(answer)
````

## Functions Can Have More Than One Parameter

A function can receive more than one value.

````python
def add(a, b):
    return a + b
````

````python
print(add(3, 5))
````

In `add(3, 5)`, `a` and `b` are the parameters, while `3` and `5` are the arguments.

## Why Are Functions Useful?

Functions help us avoid repeating code, organize programs into meaningful pieces, reuse instructions with different values, and make programs easier to read and modify.
