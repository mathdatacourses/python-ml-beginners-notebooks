## Another Way to Repeat Instructions

A `for` loop is useful when we want to go through the elements of a collection or repeat something a known number of times.

Sometimes we want repetition to continue only while a condition is `True`.

For this situation, Python provides `while`.

## Our First `while` Loop

```python
count = 1

while count <= 5:
    print(count)
    count = count + 1
```

Python prints the numbers 1 through 5.

````python
count = 1

while count <= 5:
    print(count)
    count = count + 1
````

## Reading the Loop

Python checks the condition `count <= 5` before each repetition.

If the condition is `True`, the indented block runs.

If the condition is `False`, the loop stops.

## What Happens Step by Step?

At the beginning, `count = 1`.

The condition `count <= 5` is `True`, so Python enters the loop.

## The Value Changes Inside the Loop

Inside the loop, Python first prints the current value of `count`.

Then `count = count + 1` increases `count` by 1.

This change allows the loop to make progress.

## How Does the Loop Stop?

After Python prints 5, `count` becomes 6.

Python checks `6 <= 5`.

This is `False`, so the loop stops.

## A `while` Loop Needs Three Important Parts

1. A starting value, such as `count = 1`
2. A condition, such as `while count <= 5:`
3. A change that eventually makes the condition `False`, such as `count = count + 1`

All three parts are important.

## What If We Forget to Change the Variable?

If `count` never changes, the condition may remain `True` forever.

That means the loop never stops.

## An Infinite Loop

A loop that never stops is called an **infinite loop**.

Whenever you write a `while` loop, ask:

**What will eventually make the condition False?**

We will not deliberately run an infinite loop here.

The important point is that an update such as `count = count + 1` is often necessary so the condition can eventually become `False`.

## The Condition May Be False Immediately

A `while` loop may run zero times if its condition is already `False` before the first repetition.

````python
count = 10

while count <= 5:
    print(count)
    count = count + 1
````

A `while` loop may run many times, once, or zero times.

## `for` Loop or `while` Loop?

A `for` loop is often convenient when we are working through a collection or a sequence from `range()`.

A `while` loop is useful when repetition depends directly on a condition.

## A Simple Guideline

Use a `for` loop when you are naturally working through a collection or a sequence of values.

Use a `while` loop when you want to continue while some condition remains `True`.

This is only a guideline. Some problems can be solved using either type of loop.

## What You Should Remember

- A `while` loop repeats while a condition is `True`.
- Python checks the condition before every repetition.
- A typical `while` loop has a starting value, a condition, and a change that eventually makes the condition `False`.
- If nothing causes the condition to become `False`, the program may enter an infinite loop.
- A `for` loop is often natural for collections and `range()`.
- A `while` loop is useful when repetition depends directly on a condition.
