## What If the Condition Is False?

In the previous lecture, an `if` statement ran code only when a condition was `True`.

Sometimes we want Python to do one thing when the condition is `True` and something different when it is `False`.

Python provides `else` for this situation.

## Using `if` and `else`

Suppose:

```python
score = 82
```

We can write:

```python
if score >= 70:
    print("Passed")
else:
    print("Not passed")
```

Python chooses between the two branches.

````python
score = 82

if score >= 70:
    print("Passed")
else:
    print("Not passed")
````

## Only One Branch Runs

With an `if`–`else` structure:

- if the condition is `True`, the `if` block runs,
- if the condition is `False`, the `else` block runs.

Exactly one of the two branches runs.

## A False Condition

Now use the same code with a score below 70.

````python
score = 60

if score >= 70:
    print("Passed")
else:
    print("Not passed")
````

## What If There Are More Than Two Cases?

Sometimes we need more than two possibilities.

For example:

- score at least 90 → A
- score at least 80 → B
- score at least 70 → C
- otherwise → Below C

Python provides `elif` for additional conditions.

`elif` is short for **else if**.

## Using `elif`

The general grade structure is:

````python
score = 85

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("Below C")
````

Try several values with the same structure.

````python
score = 95

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("Below C")
````

````python
score = 75

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("Below C")
````

````python
score = 65

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("Below C")
````

## Python Checks Conditions in Order

Python checks conditions from top to bottom.

As soon as it finds a `True` condition, it runs that branch and skips the remaining branches.

````python
score = 85

if score >= 90:
    print("A")
elif score >= 80:
    print("B")
elif score >= 70:
    print("C")
else:
    print("Below C")
````

## Order Matters

A score of 95 satisfies both:

```python
score >= 70
```

and:

```python
score >= 90
```

If we put the less restrictive condition first, Python stops there.

````python
score = 95

if score >= 70:
    print("C or higher")
elif score >= 90:
    print("A")
````

For score ranges, the more restrictive conditions should usually come first:

1. at least 90,
2. then at least 80,
3. then at least 70.

Remember:

**Python checks from top to bottom and stops at the first `True` condition.**

## Indentation Still Matters

Each branch contains its own indented block.

````python
temperature = 95

if temperature > 90:
    print("Hot")
else:
    print("Not hot")
````

````python
temperature = 80

if temperature > 90:
    print("Hot")
else:
    print("Not hot")
````

## A Useful Pattern

The general structure is:

```python
if condition1:
    instruction1
elif condition2:
    instruction2
else:
    instruction3
```

There can be more than one `elif` branch.

The `else` branch is optional.

Python chooses one branch from the chain.

## What You Should Remember

- Use `if` for the first condition.
- Use `elif` for additional conditions.
- Use `else` for the remaining case.
- Python checks conditions from top to bottom.
- Python stops at the first `True` condition.
- Only one branch in an `if`–`elif`–`else` chain runs.
- The order of the conditions matters.
- Indentation tells Python which instructions belong to each branch.
