## Why Do We Need Loops?

Suppose we have several scores:

```python
scores = [82, 91, 74]
```

If we want to print every score, we could write a separate `print()` statement for each element.

That works for a few values, but it becomes inconvenient when a collection contains many values.

A loop allows Python to repeat an instruction automatically.

````python
scores = [82, 91, 74]

print(scores[0])
print(scores[1])
print(scores[2])
````

## A `for` Loop

We can replace the repeated instructions with:

```python
for score in scores:
    print(score)
```

The same instruction is repeated for every value in the list.

````python
scores = [82, 91, 74]

for score in scores:
    print(score)
````

## Reading the Loop

Consider:

```python
for score in scores:
```

We can read this as:

**For each score in scores, do the following.**

Here:

- `scores` is the list,
- `score` represents one element of the list at a time,
- the colon indicates that an indented block of code follows.

## What Happens Step by Step?

Suppose:

```python
scores = [82, 91, 74]
```

Python works through the list from beginning to end:

- first `score` is 82,
- then `score` is 91,
- then `score` is 74.

After Python has processed every element, the loop ends.

## The Loop Variable Changes Automatically

In:

```python
for score in scores:
```

`score` is called the **loop variable**.

Python automatically changes its value as it moves through the list.

We do not have to change `score` ourselves.

## Indentation Tells Python What to Repeat

The indented instructions are part of the loop.

For example:

```python
for score in scores:
    print(score)
```

The line `print(score)` runs once for every value in `scores`.

Just as with If statements, indentation is part of Python syntax.

## More Than One Instruction Can Repeat

A loop can contain several indented instructions.

For every element, Python repeats the entire indented block.

````python
scores = [82, 91, 74]

for score in scores:
    print("Next score:")
    print(score)
````

## Code Outside the Loop Runs Once

A line that is not indented is outside the loop.

It runs after the loop has finished.

````python
scores = [82, 91, 74]

for score in scores:
    print(score)

print("Finished")
````

## Loops and Conditions Can Work Together

We can use an If statement inside a loop.

The loop gives us repetition.

The If statement gives us a decision.

````python
scores = [82, 91, 74]

for score in scores:
    if score >= 70:
        print(score)
````

## Choose a Descriptive Loop Variable

The name of the loop variable is our choice.

This works:

```python
for x in scores:
```

But this is easier to understand:

```python
for score in scores:
```

Similarly, for a list of temperatures, `temperature` is a useful loop variable name.

Good practice is to choose a name that describes what one element represents.

## A Common Mistake

The instructions that should repeat must be indented.

Incorrect:

```python
for score in scores:
print(score)
```

Correct:

```python
for score in scores:
    print(score)
```

Python uses indentation to identify the body of the loop.

## What You Should Remember

- A `for` loop repeats instructions for the elements of a collection.
- The collection contains the values we want to process.
- The loop variable represents one element at a time.
- The loop variable changes automatically.
- The indented block repeats for every element.
- Code outside the indented block is not repeated.
- A loop can contain more than one instruction.
- Loops can be combined with If statements.
- Choose a descriptive loop variable name when possible.
