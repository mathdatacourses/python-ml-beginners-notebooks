## So Far, Variables Have Stored One Value

For example:

```python
score = 82
name = "Anna"
```

But what if we want to store many related values?

## One Variable for Every Value?

We could create:

```python
score1 = 82
score2 = 91
score3 = 74
score4 = 88
score5 = 95
```

This works, but it quickly becomes inconvenient.

## A Collection of Values

Python provides a type called a **list**.

A list stores several values together in one object.

````python
scores = [82, 91, 74, 88, 95]
````

````python
scores
````

## Lists Use Square Brackets

List values are written inside square brackets and separated by commas.

The values inside a list are called **elements**.

## Lists Can Store Different Kinds of Values

Lists can contain numbers, strings, Boolean values, or even different types together.

````python
[10, 20, 30]
````

````python
["red", "blue", "green"]
````

````python
[True, False, True]
````

````python
[10, "Anna", True]
````

## A List Is One Python Object

We can ask Python for the type of a list.

````python
scores = [82, 91, 74, 88, 95]
````

````python
type(scores)
````

## Finding the Number of Elements

For a list, `len()` counts the number of elements.

````python
len(scores)
````

## Lists Are Ordered

The order of elements matters.

Because lists are ordered, Python can refer to elements by position.

## Lists Also Use Indices

Python begins counting list positions at **0**.

For:

```python
scores = [82, 91, 74, 88]
```

the indices are 0, 1, 2, and 3.

## Selecting an Element

Use square brackets with an index.

````python
scores = [82, 91, 74, 88]
````

````python
scores[0]
````

````python
scores[2]
````

## Why Collections Are Useful

Collections let us keep related values together and make code easier to organize.
