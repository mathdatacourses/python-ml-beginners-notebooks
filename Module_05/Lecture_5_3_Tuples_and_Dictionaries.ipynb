## More Ways to Store Collections

We already know that lists can store several values together.

Python also provides other useful collections:

- **Tuples** — ordered collections that cannot be changed after they are created
- **Dictionaries** — collections of key-value pairs

## Tuples

A tuple is similar to a list.

It stores several values in order.

Tuples are usually written using parentheses.

````python
scores = (82, 91, 74, 88)
````

````python
type(scores)
````

## Tuples Are Ordered

Like lists, tuples are ordered.

Python uses indices beginning at zero.

````python
scores[0]
````

````python
scores[2]
````

## The Main Difference: Tuples Cannot Be Changed

Lists can be changed after they are created.

Tuples cannot.

Tuples are **immutable**, which means their elements cannot be changed after the tuple is created.

````python
scores = (82, 91, 74)
````

````python
# Try removing the # from the next line to see the error:
# scores[0] = 85
````

## List or Tuple?

A list uses square brackets and can be changed.

A tuple uses parentheses and cannot be changed.

Both are ordered and use indices.

## Dictionaries Store Key-Value Pairs

A dictionary stores values using names called **keys**.

For example:

```python
student = {"name": "Anna", "score": 82}
```

The keys describe what the values represent.

## Dictionary Syntax

Dictionaries use curly braces.

Each entry contains:

```python
key: value
```

Entries are separated by commas.

````python
student = {"name": "Anna", "score": 82}
````

````python
student
````

## Accessing a Dictionary Value

To access a value in a dictionary, use its key instead of a numerical index.

````python
student["name"]
````

````python
student["score"]
````

## Changing a Dictionary Value

Dictionary values can be changed by assigning a new value to an existing key.

````python
student["score"] = 88
````

````python
student["score"]
````

## Adding a New Key-Value Pair

We can add new information by assigning a value to a new key.

````python
student["passed"] = True
````

````python
student
````

## Why Dictionaries Are Useful

Dictionaries are useful when values have meaningful names.

Using descriptive keys such as `"name"`, `"age"`, or `"score"` can make code easier to understand than remembering numerical positions.

## What You Should Remember

| Collection | Main idea | Syntax |
|---|---|---|
| List | Ordered and changeable | `[ ]` |
| Tuple | Ordered and not changeable | `( )` |
| Dictionary | Key-value pairs | `{ }` |

- Lists and tuples use positions called indices.
- Dictionaries use keys to access values.
- The best collection depends on how we want to organize the data.
