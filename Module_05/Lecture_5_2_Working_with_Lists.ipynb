## Working with a List

Suppose we have:

```python
scores = [82, 91, 74, 88, 95]
```

In the previous lecture, we learned how to select an element using its index.

Now we will build on that idea.

## Positive Indices

Python begins counting positions at zero.

````python
scores = [82, 91, 74, 88, 95]
````

````python
scores[1]
````

````python
scores[4]
````

## Negative Indices

Python can also count backward from the end of a list.

The last element has index `-1`.

The second-to-last element has index `-2`.

````python
scores[-1]
````

````python
scores[-2]
````

## Selecting Several Elements

Lists can be sliced just like strings.

Remember:

**Start included, end excluded.**

````python
scores[1:4]
````

## Leaving Out a Slice Boundary

If we leave out the starting index, Python begins at the start of the list.

If we leave out the ending index, Python continues to the end of the list.

````python
scores[:3]
````

````python
scores[2:]
````

## Changing an Element

Unlike strings, lists can be changed after they are created.

We can assign a new value to an element at a particular index.

````python
scores = [82, 91, 74]
````

````python
scores[0] = 85
````

````python
scores
````

## Adding an Element with `append()`

The method `append()` adds an element to the end of a list.

````python
scores = [82, 91, 74]
````

````python
scores.append(88)
````

````python
scores
````

## `append()` Changes the List

This is different from many string methods.

`append()` changes the existing list directly.

We do **not** need to assign the result back to the variable.

````python
scores = [82, 91, 74]
````

````python
scores.append(88)
````

````python
scores
````

## Removing an Element with `remove()`

The method `remove()` removes the first element with a specified value, if that value exists.

````python
scores = [82, 91, 74, 88]
````

````python
scores.remove(74)
````

````python
scores
````

## Removing an Element by Position

Use `del` when you want to remove an element based on its index.

````python
scores = [82, 91, 74, 88]
````

````python
del scores[1]
````

````python
scores
````

## Checking Membership with `in`

The operator `in` checks whether a value appears in a list.

The result is a Boolean value.

````python
scores = [82, 91, 74, 88]
````

````python
91 in scores
````

````python
100 in scores
````

## What You Should Remember

With Python lists, we can:

- select elements using positive or negative indices,
- select several elements using slices,
- change an element using its index,
- add an element using `append()`,
- remove a value using `remove()`,
- remove an element by position using `del`,
- check membership using `in`.

The most important new idea is that **lists can be changed after they are created**.
